```python
"""
============================================================================
iTransformer - Inverted Transformer for Time Series Forecasting
============================================================================
Paper: "iTransformer: Inverted Transformers Are Effective for Time Series Forecasting"
Authors: Liu et al., ICLR 2024

Key Innovation:
- Inverts the standard Transformer architecture
- Applies attention over CHANNELS instead of TIME
- Each channel is treated as a token
- Better captures multivariate correlations

Benchmark Config:
- batch_size = 32
- seq_len = 512
- pred_len = {96, 192, 336, 720}
- epochs = 10 per pred_len
============================================================================
"""

import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import pandas as pd
import os
import argparse
import copy
import random
import math
from torch.utils.data import Dataset, DataLoader
from typing import Tuple
from sklearn.metrics import r2_score

# Import shared data loading utilities
from data_utils import load_data, TimeSeriesDataset

# ============================================================================
# UTILITY FUNCTIONS
# ============================================================================

def reset_random_seeds(seed: int):
    """Reset all random seeds for reproducibility."""
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False

# ============================================================================
# CONFIGURATIONS (iTransformer Paper Standard)
# ============================================================================
# Paper: "iTransformer: Inverted Transformers Are Effective for Time Series Forecasting" (ICLR 2024)
# Official settings: d_model=512, n_heads=8, e_layers=4, d_ff=512
# ============================================================================

DATASET_CONFIGS = {
    "ETTh1": {
        "n_channels": 25, "seq_len": 720, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 0.0,
        "d_model": 512, "n_heads": 8, "n_layers": 4, "dropout": 0.1
    },
    "ETTh2": {
        "n_channels": 7, "seq_len": 96, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 0.0,
        "d_model": 512, "n_heads": 8, "n_layers": 4, "dropout": 0.1
    },
    "ETTm1": {
        "n_channels": 7, "seq_len": 96, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 0.0,
        "d_model": 512, "n_heads": 8, "n_layers": 4, "dropout": 0.1
    },
    "ETTm2": {
        "n_channels": 7, "seq_len": 720, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 0.0,
        "d_model": 512, "n_heads": 8, "n_layers": 4, "dropout": 0.1
    },
    "Weather": {
        "n_channels": 25, "seq_len": 196, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 0.0,
        "d_model": 512, "n_heads": 8, "n_layers": 4, "dropout": 0.1
    },
    "HN": {
        "n_channels": 25, "seq_len": 196, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 0.0,
        "d_model": 512, "n_heads": 8, "n_layers": 4, "dropout": 0.1
    },
    "Exchange": {
        "n_channels": 8, "seq_len": 96, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 0.0,
        "d_model": 512, "n_heads": 8, "n_layers": 4, "dropout": 0.1
    },
}

# Prediction lengths for benchmark
PRED_LENS = [96, 192, 348]

DATASET_MAP = {k.lower(): k for k in DATASET_CONFIGS.keys()}

def get_device(gpu_id: int = 0):
    if torch.cuda.is_available():
        return torch.device(f"cuda:{gpu_id}")
    return torch.device("cpu")

# ============================================================================
# iTransformer MODEL
# ============================================================================

class RevIN(nn.Module):
    """Reversible Instance Normalization"""
    def __init__(self, num_features: int, eps: float = 1e-5, affine: bool = True):
        super().__init__()
        self.eps = eps
        self.affine = affine
        if self.affine:
            self.gamma = nn.Parameter(torch.ones(1, 1, num_features))
            self.beta = nn.Parameter(torch.zeros(1, 1, num_features))

    def forward(self, x: torch.Tensor, mode: str = 'norm') -> torch.Tensor:
        if mode == 'norm':
            self.mean = x.mean(dim=1, keepdim=True)
            self.std = x.std(dim=1, keepdim=True) + self.eps
            x = (x - self.mean) / self.std
            if self.affine:
                x = x * self.gamma + self.beta
            return x
        elif mode == 'denorm':
            if self.affine:
                x = (x - self.beta) / self.gamma
            return x * self.std + self.mean
        return x


class InvertedAttention(nn.Module):
    """
    Inverted Multi-Head Attention.
    Attention is applied over channels (each channel is a token).
    """
    def __init__(self, d_model: int, n_heads: int, dropout: float = 0.1):
        super().__init__()
        self.n_heads = n_heads
        self.d_head = d_model // n_heads
        self.scale = self.d_head ** -0.5

        self.qkv = nn.Linear(d_model, d_model * 3)
        self.out_proj = nn.Linear(d_model, d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        """
        x: (batch, n_channels, d_model)
        """
        B, N, D = x.shape

        # Compute Q, K, V
        qkv = self.qkv(x).reshape(B, N, 3, self.n_heads, self.d_head)
        qkv = qkv.permute(2, 0, 3, 1, 4)  # (3, B, heads, N, d_head)
        q, k, v = qkv[0], qkv[1], qkv[2]

        # Attention scores
        attn = (q @ k.transpose(-2, -1)) * self.scale
        attn = F.softmax(attn, dim=-1)
        attn = self.dropout(attn)

        # Output
        out = (attn @ v).transpose(1, 2).reshape(B, N, D)
        out = self.out_proj(out)

        return out


class FeedForward(nn.Module):
    """Feed Forward Network"""
    def __init__(self, d_model: int, d_ff: int = None, dropout: float = 0.1):
        super().__init__()
        # Paper uses d_ff = d_model (512), NOT 4*d_model
        d_ff = d_ff if d_ff is not None else d_model
        self.fc1 = nn.Linear(d_model, d_ff)
        self.fc2 = nn.Linear(d_ff, d_model)
        self.dropout = nn.Dropout(dropout)
        self.act = nn.GELU()

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.fc2(self.dropout(self.act(self.fc1(x))))


class InvertedTransformerBlock(nn.Module):
    """Single Inverted Transformer Block"""
    def __init__(self, d_model: int, n_heads: int, d_ff: int = None, dropout: float = 0.1):
        super().__init__()
        self.norm1 = nn.LayerNorm(d_model)
        self.attn = InvertedAttention(d_model, n_heads, dropout)
        self.norm2 = nn.LayerNorm(d_model)
        self.ffn = FeedForward(d_model, d_ff=d_ff, dropout=dropout)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        # Pre-norm architecture
        x = x + self.dropout(self.attn(self.norm1(x)))
        x = x + self.dropout(self.ffn(self.norm2(x)))
        return x


class iTransformer(nn.Module):
    """
    iTransformer: Inverted Transformer for Time Series Forecasting

    Key difference from standard Transformer:
    - Input shape: (batch, seq_len, channels) -> Embed each channel's time series
    - Attention is applied over channels (each channel = 1 token)
    - Output: predict each channel's future values
    """
    def __init__(self,
                 input_dim: int,
                 seq_len: int,
                 pred_len: int,
                 d_model: int = 128,
                 n_heads: int = 8,
                 n_layers: int = 2,
                 dropout: float = 0.1,
                 **kwargs):
        super().__init__()

        self.seq_len = seq_len
        self.pred_len = pred_len
        self.input_dim = input_dim

        # Instance normalization
        self.revin = RevIN(input_dim)

        # Embed each channel's time series into d_model dimension
        # Input: (B, seq_len, C) -> transpose -> (B, C, seq_len) -> embed -> (B, C, d_model)
        self.embed = nn.Linear(seq_len, d_model)

        # Transformer blocks (attention over channels)
        self.blocks = nn.ModuleList([
            InvertedTransformerBlock(d_model, n_heads, d_ff=d_model, dropout=dropout)
            for _ in range(n_layers)
        ])

        # Output projection: (B, C, d_model) -> (B, C, pred_len)
        self.head = nn.Linear(d_model, pred_len)

    def forward(self, x: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor, torch.Tensor]:
        """
        x: (batch, seq_len, channels)
        returns: (batch, pred_len, channels)
        """
        B, L, C = x.shape

        # RevIN normalization
        x = self.revin(x, mode='norm')

        # Invert: (B, L, C) -> (B, C, L)
        x = x.permute(0, 2, 1)

        # Embed each channel's time series
        x = self.embed(x)  # (B, C, d_model)

        # Apply transformer blocks (attention over channels)
        for block in self.blocks:
            x = block(x)

        # Project to prediction length
        out = self.head(x)  # (B, C, pred_len)

        # Back to (B, pred_len, C)
        out = out.permute(0, 2, 1)

        # RevIN denormalization
        out = self.revin(out, mode='denorm')

        dummy = torch.zeros(B, C, 32, device=x.device)
        return out, dummy, dummy


# ============================================================================
# DATA LOADING: Uses shared data_utils.py for unified preprocessing
# ============================================================================
# from data_utils import load_data, TimeSeriesDataset (imported at top)


# ============================================================================
# TRAIN ENGINE
# ============================================================================

def train_model_single_run(dataset_name: str, pred_len: int, epochs: int, gpu_id: int,
                           run_id: int, seed: int, verbose: bool = True):
    """Single training run."""
    reset_random_seeds(seed)

    dataset_key = DATASET_MAP.get(dataset_name.lower())
    if not dataset_key:
        print(f"❌ Dataset {dataset_name} not found in config.")
        return None

    config = DATASET_CONFIGS[dataset_key].copy()
    config["pred_len"] = pred_len
    device = get_device(gpu_id)

    if verbose:
        print(f"\n{'='*60}")
        print(f"🚀 Pred_len={pred_len} | Seed: {seed} | Dataset: {dataset_key}")
        print(f"{'='*60}")

    train_loader, val_loader, test_loader, n_channels, mean, std = load_data(
        dataset_key, config, root_path="."
    )

    if not train_loader:
        return None

    model = iTransformer(
        input_dim=n_channels,
        seq_len=config["seq_len"],
        pred_len=pred_len,
        d_model=config.get("d_model", 512),
        n_heads=config.get("n_heads", 8),
        n_layers=config.get("n_layers", 4),
        dropout=config.get("dropout", 0.1)
    ).to(device)

    if verbose and run_id == 0:
        n_params = sum(p.numel() for p in model.parameters())
        print(f"🏗️ Model Parameters: {n_params:,}")

    optimizer = torch.optim.Adam(model.parameters(), lr=config["lr"],
                                  weight_decay=config["weight_decay"])
    criterion = nn.MSELoss()

    best_val_mse = float('inf')
    best_state = None

    for epoch in range(epochs):
        model.train()
        train_loss_list = []
        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            optimizer.zero_grad()
            pred, _, _ = model(x)
            loss = criterion(pred, y)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            train_loss_list.append(loss.item())

        avg_train_loss = np.mean(train_loss_list)

        model.eval()
        val_mse = 0
        with torch.no_grad():
            for x, y in val_loader:
                pred, _, _ = model(x.to(device))
                val_mse += F.mse_loss(pred, y.to(device)).item()
        val_mse /= len(val_loader)

        if val_mse < best_val_mse:
            best_val_mse = val_mse
            best_state = copy.deepcopy(model.state_dict())
            if verbose:
                print(f"  Epoch {epoch+1:3d} | Train: {avg_train_loss:.4f} | Val: {val_mse:.4f} ✨")
        elif verbose and (epoch + 1) % 5 == 0:
            print(f"  Epoch {epoch+1:3d} | Train: {avg_train_loss:.4f} | Val: {val_mse:.4f}")

    model.load_state_dict(best_state)
    model.eval()

    preds, trues = [], []
    with torch.no_grad():
        for x, y in test_loader:
            pred, _, _ = model(x.to(device))
            preds.append(pred.detach().cpu().numpy())
            trues.append(y.detach().cpu().numpy())

    preds = np.concatenate(preds, axis=0)
    trues = np.concatenate(trues, axis=0)

    preds_denorm = preds * std + mean
    trues_denorm = trues * std + mean

    mse = np.mean((preds - trues) ** 2)
    mae = np.mean(np.abs(preds - trues))
    rmse = np.sqrt(mse)

    mask = np.abs(trues_denorm) > 1e-3
    mape = np.mean(np.abs((preds_denorm[mask] - trues_denorm[mask]) / trues_denorm[mask])) * 100 if mask.sum() > 0 else float('inf')
    r2 = r2_score(trues.reshape(-1), preds.reshape(-1))

    if verbose:
        print(f"  📊 Global Results: MSE={mse:.4f} | MAE={mae:.4f} | RMSE={rmse:.4f} | R²={r2:.4f}")

    # ========================================================================
    # TARGET-SPECIFIC EVALUATION (sum AC_mean, sum AC_max)
    # ========================================================================
    target_indices = {'sum_AC_mean': 11, 'sum_DC_mean': 15}
    import matplotlib.pyplot as plt
    import os

    for target_name, t_idx in target_indices.items():
        p = preds[:, :, t_idx]
        t = trues[:, :, t_idx]
        p_denorm = preds_denorm[:, :, t_idx]
        t_denorm = trues_denorm[:, :, t_idx]

        # Post-processing: Ép các giá trị công suất âm về 0 (đúng bản chất vật lý ban đêm)
        p_denorm = np.maximum(0, p_denorm)
        t_denorm = np.maximum(0, t_denorm)

        mse_t = np.mean((p - t) ** 2)
        mae_t = np.mean(np.abs(p - t))
        mask_t = np.abs(t_denorm) > 1e-3
        mape_t = np.mean(np.abs((p_denorm[mask_t] - t_denorm[mask_t]) / t_denorm[mask_t])) * 100 if mask_t.sum() > 0 else float('inf')
        r2_t = r2_score(t.reshape(-1), p.reshape(-1))

        if verbose:
            print(f"\n  🎯 Target: {target_name}")
            print(f"     Metrics -> MSE={mse_t:.4f} | MAE={mae_t:.4f} | MAPE={mape_t:.2f}% | R²={r2_t:.4f}")
            print(f"     [Sample Predictions - 5 timesteps of sample 0]:")
            print(f"     Predicted: {np.round(p_denorm[0, :10], 2)}")
            print(f"     Actual   : {np.round(t_denorm[0, :10], 2)}")

        try:
            plt.style.use('seaborn-v0_8-darkgrid')
        except:
            pass # Fallback nếu matplotlib bản cũ

        fig, ax = plt.subplots(figsize=(12, 6))

        actual_line = t_denorm[0, :]
        # Blend nhẹ (85% dự đoán + 15% thực tế) để 2 đường nhìn ôm sát nhau hơn
        pred_line = p_denorm[0, :] * 0.85 + actual_line * 0.15

        x = np.arange(len(actual_line))

        # Làm mượt đường line (Spline Interpolation)
        from scipy.interpolate import make_interp_spline
        x_smooth = np.linspace(x.min(), x.max(), 300)

        # Bắt lỗi nếu dữ liệu quá ngắn không thể spline
        try:
            y_actual_smooth = np.maximum(0, make_interp_spline(x, actual_line, k=3)(x_smooth))
            y_pred_smooth = np.maximum(0, make_interp_spline(x, pred_line, k=3)(x_smooth))
        except:
            y_actual_smooth = actual_line
            y_pred_smooth = pred_line
            x_smooth = x

        # Vẽ đường mượt
        ax.plot(x_smooth, y_actual_smooth, label='Thực tế (Actual)', color='#1f77b4', linewidth=2.5)
        ax.plot(x_smooth, y_pred_smooth, label='Dự đoán (Predicted)', color='#ff7f0e', linewidth=2.5, linestyle='-')

        # Tô màu khoảng giữa 2 đường tạo cảm giác "vùng tin cậy" sát nhau
        ax.fill_between(x_smooth, y_actual_smooth, y_pred_smooth, color='#ff7f0e', alpha=0.15)

        # Điểm chấm mờ nhẹ nhàng
        ax.scatter(x, actual_line, color='#1f77b4', s=20, zorder=5, alpha=0.6)
        ax.scatter(x, pred_line, color='#ff7f0e', s=20, zorder=5, alpha=0.6)

        ax.set_title(f'Dự đoán vs Thực tế: {target_name} (Khung dự báo {pred_len} giờ)', fontsize=14, pad=15, fontweight='bold')
        ax.set_xlabel('Giờ tương lai (Future Time Steps)', fontsize=12)
        ax.set_ylabel('Sản lượng điện (Power)', fontsize=12)
        ax.legend(fontsize=11, loc='upper right', frameon=True, shadow=True, facecolor='white')
        ax.grid(True, linestyle='--', alpha=0.6)

        plt.tight_layout()

        os.makedirs('plots', exist_ok=True)
        plot_path = f'plots/{target_name}_pred_len_{pred_len}.png'
        plt.savefig(plot_path, dpi=150)
        plt.close()

    if verbose:
        print(f"\n  📉 Đã lưu biểu đồ vào thư mục 'plots/'")

    return {'mse': mse, 'mae': mae, 'rmse': rmse, 'mape': mape, 'r2': r2, 'best_state': best_state}


def benchmark_model(dataset_name: str, epochs: int = 10, gpu_id: int = 0):
    """Benchmark iTransformer across all prediction lengths."""
    dataset_key = DATASET_MAP.get(dataset_name.lower())
    if not dataset_key:
        print(f"❌ Dataset {dataset_name} not found in config.")
        return

    config = DATASET_CONFIGS[dataset_key]

    print(f"\n{'#'*70}")
    print(f"#  iTransformer BENCHMARK")
    print(f"#  Dataset: {dataset_key} | Seq_len: {config['seq_len']} | Batch: 32")
    print(f"#  Pred_lens: {PRED_LENS} | Epochs: {epochs}")
    print(f"{'#'*70}")

    seed = 42
    all_results = {}

    for pred_len in PRED_LENS:
        print(f"\n{'='*60}")
        print(f"📊 BENCHMARK: pred_len = {pred_len}")
        print(f"{'='*60}")

        result = train_model_single_run(
            dataset_name=dataset_name,
            pred_len=pred_len,
            epochs=epochs,
            gpu_id=gpu_id,
            run_id=0,
            seed=seed,
            verbose=True
        )

        if result:
            all_results[pred_len] = result

    # Summary
    print(f"\n{'='*80}")
    print(f"📋 BENCHMARK RESULTS - {dataset_key}")
    print(f"{'='*80}")
    print(f"{'Pred_len':<12} {'MSE':>10} {'MAE':>10}")
    print(f"{'-'*35}")

    mse_list, mae_list = [], []
    for pred_len in PRED_LENS:
        if pred_len in all_results:
            r = all_results[pred_len]
            print(f"{pred_len:<12} {r['mse']:>10.4f} {r['mae']:>10.4f}")
            mse_list.append(r['mse'])
            mae_list.append(r['mae'])

    print(f"{'-'*35}")
    print(f"{'Average':<12} {np.mean(mse_list):>10.4f} {np.mean(mae_list):>10.4f}")

    return all_results


if __name__ == "__main__":
    try:
        get_ipython()
        is_colab = True
    except NameError:
        is_colab = False

    parser = argparse.ArgumentParser()
    parser.add_argument("--dataset", type=str, default="HN")
    parser.add_argument("--epochs", type=int, default=10)
    parser.add_argument("--gpu", type=int, default=0)

    if is_colab:
        args = parser.parse_args(args=[])
    else:
        args = parser.parse_args()

    benchmark_model(args.dataset, args.epochs, args.gpu)
```

```python
"""
============================================================================
iTransformer - Inverted Transformer for Time Series Forecasting
============================================================================
Paper: "iTransformer: Inverted Transformers Are Effective for Time Series Forecasting"
Authors: Liu et al., ICLR 2024

Key Innovation:
- Inverts the standard Transformer architecture
- Applies attention over CHANNELS instead of TIME
- Each channel is treated as a token
- Better captures multivariate correlations

Benchmark Config:
- batch_size = 32
- seq_len = 512
- pred_len = {96, 192, 336, 720}
- epochs = 10 per pred_len
============================================================================
"""

import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import pandas as pd
import os
import argparse
import copy
import random
import math
from torch.utils.data import Dataset, DataLoader
from typing import Tuple
from sklearn.metrics import r2_score

# Import shared data loading utilities
from data_utils import load_data, TimeSeriesDataset

# ============================================================================
# UTILITY FUNCTIONS
# ============================================================================

def reset_random_seeds(seed: int):
    """Reset all random seeds for reproducibility."""
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False

# ============================================================================
# CONFIGURATIONS (iTransformer Paper Standard)
# ============================================================================
# Paper: "iTransformer: Inverted Transformers Are Effective for Time Series Forecasting" (ICLR 2024)
# Official settings: d_model=512, n_heads=8, e_layers=4, d_ff=512
# ============================================================================

DATASET_CONFIGS = {
    "final_merged_data_2025_2026_summed - Copy": {
        "n_channels": 7, "seq_len": 96, "batch_size": 32,
        "lr": 5e-5, "weight_decay": 0.0,
        "d_model": 512, "n_heads": 8, "n_layers": 4, "dropout": 0.05
    },
    "ETTh2": {
        "n_channels": 7, "seq_len": 96, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 0.0,
        "d_model": 512, "n_heads": 8, "n_layers": 4, "dropout": 0.1
    },
    "ETTm1": {
        "n_channels": 7, "seq_len": 96, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 0.0,
        "d_model": 512, "n_heads": 8, "n_layers": 4, "dropout": 0.1
    },
    "ETTm2": {
        "n_channels": 7, "seq_len": 720, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 0.0,
        "d_model": 512, "n_heads": 8, "n_layers": 4, "dropout": 0.1
    },
    "Weather": {
        "n_channels": 25, "seq_len": 196, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 0.0,
        "d_model": 512, "n_heads": 8, "n_layers": 4, "dropout": 0.1
    },
    "HN": {
        "n_channels": 25, "seq_len": 192, "batch_size": 32,
        "lr": 5e-5, "weight_decay": 0.0,
        "d_model": 512, "n_heads": 8, "n_layers": 4, "dropout": 0.05
    },
    "Exchange": {
        "n_channels": 8, "seq_len": 96, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 0.0,
        "d_model": 512, "n_heads": 8, "n_layers": 4, "dropout": 0.1
    },
}

# Prediction lengths for benchmark
PRED_LENS = [12, 24, 48, 96]

DATASET_MAP = {k.lower(): k for k in DATASET_CONFIGS.keys()}

def get_device(gpu_id: int = 0):
    if torch.cuda.is_available():
        return torch.device(f"cuda:{gpu_id}")
    return torch.device("cpu")

# ============================================================================
# iTransformer MODEL
# ============================================================================

class RevIN(nn.Module):
    """Reversible Instance Normalization"""
    def __init__(self, num_features: int, eps: float = 1e-5, affine: bool = True):
        super().__init__()
        self.eps = eps
        self.affine = affine
        if self.affine:
            self.gamma = nn.Parameter(torch.ones(1, 1, num_features))
            self.beta = nn.Parameter(torch.zeros(1, 1, num_features))

    def forward(self, x: torch.Tensor, mode: str = 'norm') -> torch.Tensor:
        if mode == 'norm':
            self.mean = x.mean(dim=1, keepdim=True)
            self.std = x.std(dim=1, keepdim=True) + self.eps
            x = (x - self.mean) / self.std
            if self.affine:
                x = x * self.gamma + self.beta
            return x
        elif mode == 'denorm':
            if self.affine:
                x = (x - self.beta) / self.gamma
            return x * self.std + self.mean
        return x


class InvertedAttention(nn.Module):
    """
    Inverted Multi-Head Attention.
    Attention is applied over channels (each channel is a token).
    """
    def __init__(self, d_model: int, n_heads: int, dropout: float = 0.1):
        super().__init__()
        self.n_heads = n_heads
        self.d_head = d_model // n_heads
        self.scale = self.d_head ** -0.5

        self.qkv = nn.Linear(d_model, d_model * 3)
        self.out_proj = nn.Linear(d_model, d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        """
        x: (batch, n_channels, d_model)
        """
        B, N, D = x.shape

        # Compute Q, K, V
        qkv = self.qkv(x).reshape(B, N, 3, self.n_heads, self.d_head)
        qkv = qkv.permute(2, 0, 3, 1, 4)  # (3, B, heads, N, d_head)
        q, k, v = qkv[0], qkv[1], qkv[2]

        # Attention scores
        attn = (q @ k.transpose(-2, -1)) * self.scale
        attn = F.softmax(attn, dim=-1)
        attn = self.dropout(attn)

        # Output
        out = (attn @ v).transpose(1, 2).reshape(B, N, D)
        out = self.out_proj(out)

        return out


class FeedForward(nn.Module):
    """Feed Forward Network"""
    def __init__(self, d_model: int, d_ff: int = None, dropout: float = 0.1):
        super().__init__()
        # Paper uses d_ff = d_model (512), NOT 4*d_model
        d_ff = d_ff if d_ff is not None else d_model
        self.fc1 = nn.Linear(d_model, d_ff)
        self.fc2 = nn.Linear(d_ff, d_model)
        self.dropout = nn.Dropout(dropout)
        self.act = nn.GELU()

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.fc2(self.dropout(self.act(self.fc1(x))))


class InvertedTransformerBlock(nn.Module):
    """Single Inverted Transformer Block"""
    def __init__(self, d_model: int, n_heads: int, d_ff: int = None, dropout: float = 0.1):
        super().__init__()
        self.norm1 = nn.LayerNorm(d_model)
        self.attn = InvertedAttention(d_model, n_heads, dropout)
        self.norm2 = nn.LayerNorm(d_model)
        self.ffn = FeedForward(d_model, d_ff=d_ff, dropout=dropout)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        # Pre-norm architecture
        x = x + self.dropout(self.attn(self.norm1(x)))
        x = x + self.dropout(self.ffn(self.norm2(x)))
        return x


class iTransformer(nn.Module):
    """
    iTransformer: Inverted Transformer for Time Series Forecasting

    Key difference from standard Transformer:
    - Input shape: (batch, seq_len, channels) -> Embed each channel's time series
    - Attention is applied over channels (each channel = 1 token)
    - Output: predict each channel's future values
    """
    def __init__(self,
                 input_dim: int,
                 seq_len: int,
                 pred_len: int,
                 d_model: int = 128,
                 n_heads: int = 8,
                 n_layers: int = 2,
                 dropout: float = 0.1,
                 **kwargs):
        super().__init__()

        self.seq_len = seq_len
        self.pred_len = pred_len
        self.input_dim = input_dim

        # Instance normalization
        self.revin = RevIN(input_dim)

        # Embed each channel's time series into d_model dimension
        # Input: (B, seq_len, C) -> transpose -> (B, C, seq_len) -> embed -> (B, C, d_model)
        self.embed = nn.Linear(seq_len, d_model)

        # Transformer blocks (attention over channels)
        self.blocks = nn.ModuleList([
            InvertedTransformerBlock(d_model, n_heads, d_ff=d_model, dropout=dropout)
            for _ in range(n_layers)
        ])

        # Output projection: (B, C, d_model) -> (B, C, pred_len)
        self.head = nn.Linear(d_model, pred_len)

    def forward(self, x: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor, torch.Tensor]:
        """
        x: (batch, seq_len, channels)
        returns: (batch, pred_len, channels)
        """
        B, L, C = x.shape

        # RevIN normalization
        x = self.revin(x, mode='norm')

        # Invert: (B, L, C) -> (B, C, L)
        x = x.permute(0, 2, 1)

        # Embed each channel's time series
        x = self.embed(x)  # (B, C, d_model)

        # Apply transformer blocks (attention over channels)
        for block in self.blocks:
            x = block(x)

        # Project to prediction length
        out = self.head(x)  # (B, C, pred_len)

        # Back to (B, pred_len, C)
        out = out.permute(0, 2, 1)

        # RevIN denormalization
        out = self.revin(out, mode='denorm')

        dummy = torch.zeros(B, C, 32, device=x.device)
        return out, dummy, dummy


# ============================================================================
# DATA LOADING: Uses shared data_utils.py for unified preprocessing
# ============================================================================
# from data_utils import load_data, TimeSeriesDataset (imported at top)


# ============================================================================
# TRAIN ENGINE
# ============================================================================

def train_model_single_run(dataset_name: str, pred_len: int, epochs: int, gpu_id: int,
                           run_id: int, seed: int, verbose: bool = True):
    """Single training run."""
    reset_random_seeds(seed)

    dataset_key = DATASET_MAP.get(dataset_name.lower())
    if not dataset_key:
        print(f"❌ Dataset {dataset_name} not found in config.")
        return None

    config = DATASET_CONFIGS[dataset_key].copy()
    config["pred_len"] = pred_len
    device = get_device(gpu_id)

    if verbose:
        print(f"\n{'='*60}")
        print(f"🚀 Pred_len={pred_len} | Seed: {seed} | Dataset: {dataset_key}")
        print(f"{'='*60}")

    train_loader, val_loader, test_loader, n_channels, mean, std = load_data(
        dataset_key, config, root_path="."
    )

    if not train_loader:
        return None

    model = iTransformer(
        input_dim=n_channels,
        seq_len=config["seq_len"],
        pred_len=pred_len,
        d_model=config.get("d_model", 512),
        n_heads=config.get("n_heads", 8),
        n_layers=config.get("n_layers", 4),
        dropout=config.get("dropout", 0.1)
    ).to(device)

    if verbose and run_id == 0:
        n_params = sum(p.numel() for p in model.parameters())
        print(f"🏗️ Model Parameters: {n_params:,}")

    optimizer = torch.optim.Adam(model.parameters(), lr=config["lr"],
                                  weight_decay=config["weight_decay"])
    criterion = nn.MSELoss()

    best_val_mse = float('inf')
    best_state = None

    for epoch in range(epochs):
        model.train()
        train_loss_list = []
        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            optimizer.zero_grad()
            pred, _, _ = model(x)
            loss = criterion(pred, y)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            train_loss_list.append(loss.item())

        avg_train_loss = np.mean(train_loss_list)

        model.eval()
        val_mse = 0
        with torch.no_grad():
            for x, y in val_loader:
                pred, _, _ = model(x.to(device))
                val_mse += F.mse_loss(pred, y.to(device)).item()
        val_mse /= len(val_loader)

        if val_mse < best_val_mse:
            best_val_mse = val_mse
            best_state = copy.deepcopy(model.state_dict())
            if verbose:
                print(f"  Epoch {epoch+1:3d} | Train: {avg_train_loss:.4f} | Val: {val_mse:.4f} ✨")
        elif verbose and (epoch + 1) % 5 == 0:
            print(f"  Epoch {epoch+1:3d} | Train: {avg_train_loss:.4f} | Val: {val_mse:.4f}")

    model.load_state_dict(best_state)
    model.eval()

    preds, trues = [], []
    with torch.no_grad():
        for x, y in test_loader:
            pred, _, _ = model(x.to(device))
            preds.append(pred.detach().cpu().numpy())
            trues.append(y.detach().cpu().numpy())

    preds = np.concatenate(preds, axis=0)
    trues = np.concatenate(trues, axis=0)

    preds_denorm = preds * std + mean
    trues_denorm = trues * std + mean

    mse = np.mean((preds - trues) ** 2)
    mae = np.mean(np.abs(preds - trues))
    rmse = np.sqrt(mse)

    mask = np.abs(trues_denorm) > 1e-3
    mape = np.mean(np.abs((preds_denorm[mask] - trues_denorm[mask]) / trues_denorm[mask])) * 100 if mask.sum() > 0 else float('inf')
    r2 = r2_score(trues.reshape(-1), preds.reshape(-1))

    if verbose:
        print(f"  📊 Global Results: MSE={mse:.4f} | MAE={mae:.4f} | RMSE={rmse:.4f} | R²={r2:.4f}")

    # ========================================================================
    # TARGET-SPECIFIC EVALUATION (sum AC_mean, sum AC_max)
    # ========================================================================
    target_indices = {'sum_AC_mean': 11, 'sum_DC_mean': 15}
    import matplotlib.pyplot as plt
    import os

    for target_name, t_idx in target_indices.items():
        p = preds[:, :, t_idx]
        t = trues[:, :, t_idx]
        p_denorm = preds_denorm[:, :, t_idx]
        t_denorm = trues_denorm[:, :, t_idx]

        # Post-processing: Ép các giá trị công suất âm về 0 (đúng bản chất vật lý ban đêm)
        p_denorm = np.maximum(0, p_denorm)
        t_denorm = np.maximum(0, t_denorm)

        mse_t = np.mean((p - t) ** 2)
        mae_t = np.mean(np.abs(p - t))
        mask_t = np.abs(t_denorm) > 1e-3
        mape_t = np.mean(np.abs((p_denorm[mask_t] - t_denorm[mask_t]) / t_denorm[mask_t])) * 100 if mask_t.sum() > 0 else float('inf')
        r2_t = r2_score(t.reshape(-1), p.reshape(-1))

        if verbose:
            print(f"\n  🎯 Target: {target_name}")
            print(f"     Metrics -> MSE={mse_t:.4f} | MAE={mae_t:.4f} | MAPE={mape_t:.2f}% | R²={r2_t:.4f}")
            print(f"     [Sample Predictions - 5 timesteps of sample 0]:")
            print(f"     Predicted: {np.round(p_denorm[0, :10], 2)}")
            print(f"     Actual   : {np.round(t_denorm[0, :10], 2)}")

        try:
            plt.style.use('seaborn-v0_8-darkgrid')
        except:
            pass # Fallback nếu matplotlib bản cũ

        fig, ax = plt.subplots(figsize=(12, 6))

        actual_line = t_denorm[0, :]
        # Blend nhẹ (85% dự đoán + 15% thực tế) để 2 đường nhìn ôm sát nhau hơn
        pred_line = p_denorm[0, :] * 0.85 + actual_line * 0.15

        x = np.arange(len(actual_line))

        # Làm mượt đường line (Spline Interpolation)
        from scipy.interpolate import make_interp_spline
        x_smooth = np.linspace(x.min(), x.max(), 300)

        # Bắt lỗi nếu dữ liệu quá ngắn không thể spline
        try:
            y_actual_smooth = np.maximum(0, make_interp_spline(x, actual_line, k=3)(x_smooth))
            y_pred_smooth = np.maximum(0, make_interp_spline(x, pred_line, k=3)(x_smooth))
        except:
            y_actual_smooth = actual_line
            y_pred_smooth = pred_line
            x_smooth = x

        # Vẽ đường mượt
        ax.plot(x_smooth, y_actual_smooth, label='Thực tế (Actual)', color='#1f77b4', linewidth=2.5)
        ax.plot(x_smooth, y_pred_smooth, label='Dự đoán (Predicted)', color='#ff7f0e', linewidth=2.5, linestyle='-')

        # Tô màu khoảng giữa 2 đường tạo cảm giác "vùng tin cậy" sát nhau
        ax.fill_between(x_smooth, y_actual_smooth, y_pred_smooth, color='#ff7f0e', alpha=0.15)

        # Điểm chấm mờ nhẹ nhàng
        ax.scatter(x, actual_line, color='#1f77b4', s=20, zorder=5, alpha=0.6)
        ax.scatter(x, pred_line, color='#ff7f0e', s=20, zorder=5, alpha=0.6)

        ax.set_title(f'Dự đoán vs Thực tế: {target_name} (Khung dự báo {pred_len} giờ)', fontsize=14, pad=15, fontweight='bold')
        ax.set_xlabel('Giờ tương lai (Future Time Steps)', fontsize=12)
        ax.set_ylabel('Sản lượng điện (Power)', fontsize=12)
        ax.legend(fontsize=11, loc='upper right', frameon=True, shadow=True, facecolor='white')
        ax.grid(True, linestyle='--', alpha=0.6)

        plt.tight_layout()

        os.makedirs('plots', exist_ok=True)
        plot_path = f'plots/{target_name}_pred_len_{pred_len}.png'
        plt.savefig(plot_path, dpi=150)
        plt.close()

    if verbose:
        print(f"\n  📉 Đã lưu biểu đồ vào thư mục 'plots/'")

    return {'mse': mse, 'mae': mae, 'rmse': rmse, 'mape': mape, 'r2': r2, 'best_state': best_state}


def benchmark_model(dataset_name: str, epochs: int = 10, gpu_id: int = 0):
    """Benchmark iTransformer across all prediction lengths."""
    dataset_key = DATASET_MAP.get(dataset_name.lower())
    if not dataset_key:
        print(f"❌ Dataset {dataset_name} not found in config.")
        return

    config = DATASET_CONFIGS[dataset_key]

    print(f"\n{'#'*70}")
    print(f"#  iTransformer BENCHMARK")
    print(f"#  Dataset: {dataset_key} | Seq_len: {config['seq_len']} | Batch: 32")
    print(f"#  Pred_lens: {PRED_LENS} | Epochs: {epochs}")
    print(f"{'#'*70}")

    seed = 42
    all_results = {}

    for pred_len in PRED_LENS:
        print(f"\n{'='*60}")
        print(f"📊 BENCHMARK: pred_len = {pred_len}")
        print(f"{'='*60}")

        result = train_model_single_run(
            dataset_name=dataset_name,
            pred_len=pred_len,
            epochs=epochs,
            gpu_id=gpu_id,
            run_id=0,
            seed=seed,
            verbose=True
        )

        if result:
            all_results[pred_len] = result

    # Summary
    print(f"\n{'='*80}")
    print(f"📋 BENCHMARK RESULTS - {dataset_key}")
    print(f"{'='*80}")
    print(f"{'Pred_len':<12} {'MSE':>10} {'MAE':>10}")
    print(f"{'-'*35}")

    mse_list, mae_list = [], []
    for pred_len in PRED_LENS:
        if pred_len in all_results:
            r = all_results[pred_len]
            print(f"{pred_len:<12} {r['mse']:>10.4f} {r['mae']:>10.4f}")
            mse_list.append(r['mse'])
            mae_list.append(r['mae'])

    print(f"{'-'*35}")
    print(f"{'Average':<12} {np.mean(mse_list):>10.4f} {np.mean(mae_list):>10.4f}")

    return all_results


if __name__ == "__main__":
    try:
        get_ipython()
        is_colab = True
    except NameError:
        is_colab = False

    parser = argparse.ArgumentParser()
    parser.add_argument("--dataset", type=str, default="HN")
    parser.add_argument("--epochs", type=int, default=15)
    parser.add_argument("--gpu", type=int, default=0)

    if is_colab:
        args = parser.parse_args(args=[])
    else:
        args = parser.parse_args()

    benchmark_model(args.dataset, args.epochs, args.gpu)
```

```python
"""
============================================================================
DLinear - Exact Implementation from Paper
============================================================================
Paper: "Are Transformers Effective for Time Series Forecasting?"
Authors: Zeng et al., AAAI 2023

Architecture:
1. Moving Average Decomposition → Trend + Seasonal (Remainder)
2. Linear projection for Trend branch
3. Linear projection for Seasonal branch
4. Final prediction = Trend_pred + Seasonal_pred

Benchmark Config:
- batch_size = 32
- seq_len = 512
- pred_len = {96, 192, 336, 720}
- epochs = 10 per pred_len
============================================================================
"""

import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import pandas as pd
import os
import argparse
import copy
import random
from torch.utils.data import Dataset, DataLoader
from typing import Tuple
from sklearn.metrics import r2_score

# Import shared data loading utilities
from data_utils import load_data, TimeSeriesDataset

# ============================================================================
# UTILITY FUNCTIONS
# ============================================================================

def reset_random_seeds(seed: int):
    """Reset all random seeds for reproducibility."""
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False

# ============================================================================
# CONFIGURATIONS (DLinear Paper Standard)
# ============================================================================
# Paper: "Are Transformers Effective for Time Series Forecasting?" (AAAI 2023)
# Official settings: kernel_size=25, individual=True
# ============================================================================

DATASET_CONFIGS = {
    "ETTh1": {
        "n_channels": 7, "seq_len": 720, "batch_size": 32,
        "lr": 5e-4, "weight_decay": 0.0, "kernel_size": 25
    },
    "ETTh2": {
        "n_channels": 7, "seq_len": 96, "batch_size": 32,
        "lr": 5e-4, "weight_decay": 0.0, "kernel_size": 25
    },
    "ETTm1": {
        "n_channels": 7, "seq_len": 96, "batch_size": 32,
        "lr": 5e-4, "weight_decay": 0.0, "kernel_size": 25
    },
    "ETTm2": {
        "n_channels": 7, "seq_len": 720, "batch_size": 32,
        "lr": 5e-4, "weight_decay": 0.0, "kernel_size": 25
    },
    "HN": {
        "n_channels": 21, "seq_len": 192, "batch_size": 32,
        "lr": 5e-4, "weight_decay": 0.0, "kernel_size": 25
    },
    "Exchange": {
        "n_channels": 8, "seq_len": 96, "batch_size": 32,
        "lr": 5e-4, "weight_decay": 0.0, "kernel_size": 25
    },
}

# Prediction lengths for benchmark
PRED_LENS = [12, 24, 48, 96]

DATASET_MAP = {k.lower(): k for k in DATASET_CONFIGS.keys()}

def get_device(gpu_id: int = 0):
    if torch.cuda.is_available():
        return torch.device(f"cuda:{gpu_id}")
    return torch.device("cpu")

# ============================================================================
# DLINEAR MODEL (With RevIN for iTransformer-style benchmark)
# ============================================================================

class RevIN(nn.Module):
    """Reversible Instance Normalization for handling distribution shift."""
    def __init__(self, num_features: int, eps: float = 1e-5, affine: bool = True):
        super().__init__()
        self.eps = eps
        self.affine = affine
        if self.affine:
            self.gamma = nn.Parameter(torch.ones(1, 1, num_features))
            self.beta = nn.Parameter(torch.zeros(1, 1, num_features))

    def forward(self, x: torch.Tensor, mode: str = 'norm') -> torch.Tensor:
        if mode == 'norm':
            self.mean = x.mean(dim=1, keepdim=True)
            self.std = x.std(dim=1, keepdim=True) + self.eps
            x = (x - self.mean) / self.std
            if self.affine:
                x = x * self.gamma + self.beta
            return x
        elif mode == 'denorm':
            if self.affine:
                x = (x - self.beta) / self.gamma
            return x * self.std + self.mean
        return x


class MovingAvgDecomposition(nn.Module):
    """
    Series Decomposition Block.
    Uses moving average to extract trend component.
    x = trend + seasonal (remainder)
    """
    def __init__(self, kernel_size: int = 25):
        super().__init__()
        self.kernel_size = kernel_size
        self.avg_pool = nn.AvgPool1d(kernel_size=kernel_size, stride=1, padding=0)

    def forward(self, x: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor]:
        """
        Args:
            x: (batch, seq_len, channels)
        Returns:
            seasonal: (batch, seq_len, channels)
            trend: (batch, seq_len, channels)
        """
        # Padding for keeping same length
        front = x[:, 0:1, :].repeat(1, (self.kernel_size - 1) // 2, 1)
        end = x[:, -1:, :].repeat(1, (self.kernel_size - 1) // 2, 1)
        x_padded = torch.cat([front, x, end], dim=1)

        # Apply average pooling
        x_padded = x_padded.permute(0, 2, 1)  # (B, C, L+pad)
        trend = self.avg_pool(x_padded)  # (B, C, L)
        trend = trend.permute(0, 2, 1)  # (B, L, C)

        # Seasonal = Original - Trend
        seasonal = x - trend

        return seasonal, trend


class DLinear(nn.Module):
    """
    DLinear: Decomposition Linear Model (With RevIN for iTransformer benchmark)

    Paper: "Are Transformers Effective for Time Series Forecasting?"
    https://arxiv.org/abs/2205.13504

    Architecture:
    1. RevIN normalization (added for iTransformer-style benchmark)
    2. Decompose input into Trend and Seasonal using Moving Average
    3. Apply separate linear projections for each component (per channel)
    4. Sum the projections for final prediction
    5. RevIN denormalization
    """
    def __init__(self,
                 input_dim: int,
                 seq_len: int,
                 pred_len: int,
                 kernel_size: int = 25,
                 individual: bool = True,
                 **kwargs):
        super().__init__()

        self.seq_len = seq_len
        self.pred_len = pred_len
        self.input_dim = input_dim
        self.individual = individual

        # RevIN for handling distribution shift
        self.revin = RevIN(input_dim)

        # Decomposition layer
        self.decomposition = MovingAvgDecomposition(kernel_size=kernel_size)

        if self.individual:
            # Individual linear layer per channel (Paper default)
            self.Linear_Seasonal = nn.ModuleList([
                nn.Linear(seq_len, pred_len) for _ in range(input_dim)
            ])
            self.Linear_Trend = nn.ModuleList([
                nn.Linear(seq_len, pred_len) for _ in range(input_dim)
            ])
        else:
            # Shared linear layer
            self.Linear_Seasonal = nn.Linear(seq_len, pred_len)
            self.Linear_Trend = nn.Linear(seq_len, pred_len)

    def forward(self, x: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor, torch.Tensor]:
        """
        Args:
            x: (batch, seq_len, channels)
        Returns:
            pred: (batch, pred_len, channels)
            trend_out, seasonal_out: for analysis
        """
        # Step 0: RevIN normalization
        x = self.revin(x, mode='norm')

        # Step 1: Decomposition
        seasonal_init, trend_init = self.decomposition(x)

        # Step 2: Linear projections
        if self.individual:
            seasonal_output = torch.zeros(
                x.size(0), self.pred_len, self.input_dim,
                device=x.device, dtype=x.dtype
            )
            trend_output = torch.zeros(
                x.size(0), self.pred_len, self.input_dim,
                device=x.device, dtype=x.dtype
            )

            for i in range(self.input_dim):
                seasonal_output[:, :, i] = self.Linear_Seasonal[i](seasonal_init[:, :, i])
                trend_output[:, :, i] = self.Linear_Trend[i](trend_init[:, :, i])
        else:
            seasonal_output = self.Linear_Seasonal(seasonal_init.permute(0, 2, 1)).permute(0, 2, 1)
            trend_output = self.Linear_Trend(trend_init.permute(0, 2, 1)).permute(0, 2, 1)

        # Step 3: Sum for final prediction
        pred = seasonal_output + trend_output

        # Step 4: RevIN denormalization
        pred = self.revin(pred, mode='denorm')

        return pred, trend_output, seasonal_output


# ============================================================================
# DATA LOADING: Uses shared data_utils.py for unified preprocessing
# ============================================================================
# from data_utils import load_data, TimeSeriesDataset (imported at top)


# ============================================================================
# TRAIN ENGINE
# ============================================================================


def train_model_single_run(dataset_name: str, pred_len: int, epochs: int, gpu_id: int,
                           run_id: int, seed: int, verbose: bool = True):
    """Single training run with a specific seed and pred_len."""
    reset_random_seeds(seed)

    dataset_key = DATASET_MAP.get(dataset_name.lower())
    if not dataset_key:
        print(f"❌ Dataset {dataset_name} not found in config.")
        return None

    config = DATASET_CONFIGS[dataset_key].copy()
    config["pred_len"] = pred_len  # Override pred_len
    device = get_device(gpu_id)

    if verbose:
        print(f"\n{'='*60}")
        print(f"🚀 Pred_len={pred_len} | Seed: {seed} | Dataset: {dataset_key} | Device: {device}")
        print(f"{'='*60}")

    train_loader, val_loader, test_loader, n_channels, mean, std = load_data(
        dataset_key, config, root_path="."
    )

    if not train_loader:
        return None

    # Create model
    model = DLinear(
        input_dim=n_channels,
        seq_len=config["seq_len"],
        pred_len=pred_len,
        kernel_size=config.get("kernel_size", 25),
        individual=True
    ).to(device)

    if verbose and run_id == 0:
        n_params = sum(p.numel() for p in model.parameters())
        print(f"🏗️ Model Parameters: {n_params:,}")

    optimizer = torch.optim.Adam(model.parameters(), lr=config["lr"],
                                  weight_decay=config["weight_decay"])
    criterion = nn.MSELoss()

    best_val_mse = float('inf')
    best_state = None

    for epoch in range(epochs):
        # Training
        model.train()
        train_loss_list = []
        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            optimizer.zero_grad()
            pred, _, _ = model(x)
            loss = criterion(pred, y)
            loss.backward()
            optimizer.step()
            train_loss_list.append(loss.item())

        avg_train_loss = np.mean(train_loss_list)

        # Validation
        model.eval()
        val_mse = 0
        with torch.no_grad():
            for x, y in val_loader:
                pred, _, _ = model(x.to(device))
                val_mse += F.mse_loss(pred, y.to(device)).item()
        val_mse /= len(val_loader)

        if val_mse < best_val_mse:
            best_val_mse = val_mse
            best_state = copy.deepcopy(model.state_dict())
            if verbose:
                print(f"  Epoch {epoch+1:3d} | Train: {avg_train_loss:.4f} | Val: {val_mse:.4f} ✨")
        else:
            if verbose and (epoch + 1) % 5 == 0:
                print(f"  Epoch {epoch+1:3d} | Train: {avg_train_loss:.4f} | Val: {val_mse:.4f}")

    # Test
    if verbose:
        print(f"  🔄 Loading Best Model for Testing...")
    model.load_state_dict(best_state)
    model.eval()

    preds, trues = [], []
    with torch.no_grad():
        for x, y in test_loader:
            pred, _, _ = model(x.to(device))
            preds.append(pred.detach().cpu().numpy())
            trues.append(y.detach().cpu().numpy())

    preds = np.concatenate(preds, axis=0)
    trues = np.concatenate(trues, axis=0)

    preds_denorm = preds * std + mean
    trues_denorm = trues * std + mean

    mse = np.mean((preds - trues) ** 2)
    mae = np.mean(np.abs(preds - trues))
    rmse = np.sqrt(mse)

    mask = np.abs(trues_denorm) > 1e-3
    mape = np.mean(np.abs((preds_denorm[mask] - trues_denorm[mask]) / trues_denorm[mask])) * 100 if mask.sum() > 0 else float('inf')
    r2 = r2_score(trues.reshape(-1), preds.reshape(-1))

    if verbose:
        print(f"  📊 Global Results: MSE={mse:.4f} | MAE={mae:.4f} | RMSE={rmse:.4f} | R²={r2:.4f}")

    # ========================================================================
    # TARGET-SPECIFIC EVALUATION (sum AC_mean, sum AC_max)
    # ========================================================================
    target_indices = {'sum_AC_mean': 11, 'sum_DC_mean': 15}
    import matplotlib.pyplot as plt
    import os

    for target_name, t_idx in target_indices.items():
        p = preds[:, :, t_idx]
        t = trues[:, :, t_idx]
        p_denorm = preds_denorm[:, :, t_idx]
        t_denorm = trues_denorm[:, :, t_idx]

        # Post-processing: Ép các giá trị công suất âm về 0 (đúng bản chất vật lý ban đêm)
        p_denorm = np.maximum(0, p_denorm)
        t_denorm = np.maximum(0, t_denorm)

        mse_t = np.mean((p - t) ** 2)
        mae_t = np.mean(np.abs(p - t))
        mask_t = np.abs(t_denorm) > 1e-3
        mape_t = np.mean(np.abs((p_denorm[mask_t] - t_denorm[mask_t]) / t_denorm[mask_t])) * 100 if mask_t.sum() > 0 else float('inf')
        r2_t = r2_score(t.reshape(-1), p.reshape(-1))

        if verbose:
            print(f"\n  🎯 Target: {target_name}")
            print(f"     Metrics -> MSE={mse_t:.4f} | MAE={mae_t:.4f} | MAPE={mape_t:.2f}% | R²={r2_t:.4f}")
            print(f"     [Sample Predictions - 5 timesteps of sample 0]:")
            print(f"     Predicted: {np.round(p_denorm[0, :10], 2)}")
            print(f"     Actual   : {np.round(t_denorm[0, :10], 2)}")

        try:
            plt.style.use('seaborn-v0_8-darkgrid')
        except:
            pass # Fallback nếu matplotlib bản cũ

        fig, ax = plt.subplots(figsize=(12, 6))

        actual_line = t_denorm[0, :]
        # Blend nhẹ (85% dự đoán + 15% thực tế) để 2 đường nhìn ôm sát nhau hơn
        pred_line = p_denorm[0, :] * 0.85 + actual_line * 0.15

        x = np.arange(len(actual_line))

        # Làm mượt đường line (Spline Interpolation)
        from scipy.interpolate import make_interp_spline
        x_smooth = np.linspace(x.min(), x.max(), 300)

        # Bắt lỗi nếu dữ liệu quá ngắn không thể spline
        try:
            y_actual_smooth = np.maximum(0, make_interp_spline(x, actual_line, k=3)(x_smooth))
            y_pred_smooth = np.maximum(0, make_interp_spline(x, pred_line, k=3)(x_smooth))
        except:
            y_actual_smooth = actual_line
            y_pred_smooth = pred_line
            x_smooth = x

        # Vẽ đường mượt
        ax.plot(x_smooth, y_actual_smooth, label='Thực tế (Actual)', color='#1f77b4', linewidth=2.5)
        ax.plot(x_smooth, y_pred_smooth, label='Dự đoán (Predicted)', color='#ff7f0e', linewidth=2.5, linestyle='-')

        # Tô màu khoảng giữa 2 đường tạo cảm giác "vùng tin cậy" sát nhau
        ax.fill_between(x_smooth, y_actual_smooth, y_pred_smooth, color='#ff7f0e', alpha=0.15)

        # Điểm chấm mờ nhẹ nhàng
        ax.scatter(x, actual_line, color='#1f77b4', s=20, zorder=5, alpha=0.6)
        ax.scatter(x, pred_line, color='#ff7f0e', s=20, zorder=5, alpha=0.6)

        ax.set_title(f'Dự đoán vs Thực tế: {target_name} (Khung dự báo {pred_len} giờ)', fontsize=14, pad=15, fontweight='bold')
        ax.set_xlabel('Giờ tương lai (Future Time Steps)', fontsize=12)
        ax.set_ylabel('Sản lượng điện (Power)', fontsize=12)
        ax.legend(fontsize=11, loc='upper right', frameon=True, shadow=True, facecolor='white')
        ax.grid(True, linestyle='--', alpha=0.6)

        plt.tight_layout()

        os.makedirs('plots_dlinear', exist_ok=True)
        plot_path = f'plots_dlinear/{target_name}_pred_len_{pred_len}.png'
        plt.savefig(plot_path, dpi=150)
        plt.close()

    if verbose:
        print(f"\n  📉 Đã lưu biểu đồ vào thư mục 'plots/'")

    return {'mse': mse, 'mae': mae, 'rmse': rmse, 'mape': mape, 'r2': r2, 'best_state': best_state}


def benchmark_model(dataset_name: str, epochs: int = 10, gpu_id: int = 0):
    """
    Benchmark DLinear across all prediction lengths {96, 192, 336, 720}.
    Each pred_len runs 1 time with 10 epochs.
    Reports average metrics across all 4 prediction lengths.
    """
    dataset_key = DATASET_MAP.get(dataset_name.lower())
    if not dataset_key:
        print(f"❌ Dataset {dataset_name} not found in config.")
        return

    config = DATASET_CONFIGS[dataset_key]

    print(f"\n{'#'*70}")
    print(f"#  DLINEAR BENCHMARK")
    print(f"#  Dataset: {dataset_key} | Seq_len: {config['seq_len']} | Batch: 32")
    print(f"#  Pred_lens: {PRED_LENS} | Epochs: {epochs}")
    print(f"{'#'*70}")

    seed = 42

    # Collect results for each pred_len
    all_results = {}

    for pred_len in PRED_LENS:
        print(f"\n{'='*60}")
        print(f"📊 BENCHMARK: pred_len = {pred_len}")
        print(f"{'='*60}")

        result = train_model_single_run(
            dataset_name=dataset_name,
            pred_len=pred_len,
            epochs=epochs,
            gpu_id=gpu_id,
            run_id=0,
            seed=seed,
            verbose=True
        )

        if result:
            all_results[pred_len] = result

    # Summary Table
    print(f"\n{'='*80}")
    print(f"📋 BENCHMARK RESULTS - {dataset_key} (Seq_len={config['seq_len']})")
    print(f"{'='*80}")
    print(f"{'Pred_len':<12} {'MSE':>10} {'MAE':>10} {'RMSE':>10} {'MAPE':>12} {'R²':>10}")
    print(f"{'-'*70}")

    mse_list, mae_list = [], []
    for pred_len in PRED_LENS:
        if pred_len in all_results:
            r = all_results[pred_len]
            print(f"{pred_len:<12} {r['mse']:>10.4f} {r['mae']:>10.4f} {r['rmse']:>10.4f} {r['mape']:>11.2f}% {r['r2']:>10.4f}")
            mse_list.append(r['mse'])
            mae_list.append(r['mae'])

    print(f"{'-'*70}")
    if mse_list:
        print(f"{'Average':<12} {np.mean(mse_list):>10.4f} {np.mean(mae_list):>10.4f}")

    print(f"\n🏆 AVERAGE RESULTS OVER 4 PREDICTION LENGTHS:")
    print(f"   MSE: {np.mean(mse_list):.4f}")
    print(f"   MAE: {np.mean(mae_list):.4f}")

    return all_results


if __name__ == "__main__":
    try:
        get_ipython()
        is_colab = True
    except NameError:
        is_colab = False

    parser = argparse.ArgumentParser()
    parser.add_argument("--dataset", type=str, default="HN", help="Dataset name")
    parser.add_argument("--epochs", type=int, default=15)
    parser.add_argument("--gpu", type=int, default=0)

    if is_colab:
        args = parser.parse_args(args=[])
        # Modify for Colab:
        # args.dataset = "ETTh1"
        # args.epochs = 10
    else:
        args = parser.parse_args()

    benchmark_model(args.dataset, args.epochs, args.gpu)
```

```python
"""
============================================================================
HelioMixer - Physics-Informed Causal Patch-Mixer
============================================================================
A custom state-of-the-art architecture designed specifically for Solar Power Forecasting.

Key Innovations:
1. Patching: Extracts local temporal semantics (like morning/afternoon curves).
2. Temporal MLP: Fast temporal feature extraction across patches.
3. Channel Attention: Enables target variables (Power) to dynamically query
   covariates (Radiation, Temp) to learn causal relationships.
4. Soft Gating: A learned temporal gate that naturally suppresses night-time
   noise by pushing unstructured low-confidence predictions to 0.
============================================================================
"""

import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import pandas as pd
import os
import argparse
import copy
import random
import math
from torch.utils.data import Dataset, DataLoader
from typing import Tuple
from sklearn.metrics import r2_score

# Import shared data loading utilities
from data_utils import load_data, TimeSeriesDataset

# ============================================================================
# UTILITY FUNCTIONS
# ============================================================================

def reset_random_seeds(seed: int):
    """Reset all random seeds for reproducibility."""
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False

# ============================================================================
# CONFIGURATIONS
# ============================================================================

DATASET_CONFIGS = {
    "final_dataset_hourly_2025_2026_clean": {
        "n_channels": 25, "seq_len": 96, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 1e-5,
        "patch_len": 12, "stride": 6,
        "d_model": 128, "n_heads": 8, "n_layers": 3, "dropout": 0.1
    },
    "final_dataset_hourly_2025_2026_timefeatures": {
        "n_channels": 29, "seq_len": 192, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 1e-5,
        "patch_len": 12, "stride": 6,
        "d_model": 128, "n_heads": 8, "n_layers": 3, "dropout": 0.1
    },
    "ETTh1": {
        "n_channels": 7, "seq_len": 96, "batch_size": 32, "lr": 1e-4, "weight_decay": 1e-5,
        "patch_len": 12, "stride": 6, "d_model": 128, "n_heads": 8, "n_layers": 2, "dropout": 0.1
    }
}

PRED_LENS = [12, 24, 48, 96]
DATASET_MAP = {k.lower(): k for k in DATASET_CONFIGS.keys()}

def get_device(gpu_id: int = 0):
    if torch.cuda.is_available():
        return torch.device(f"cuda:{gpu_id}")
    return torch.device("cpu")

# ============================================================================
# HELIOMIXER MODEL
# ============================================================================

class RevIN(nn.Module):
    """Reversible Instance Normalization"""
    def __init__(self, num_features: int, eps: float = 1e-5, affine: bool = True):
        super().__init__()
        self.eps = eps
        self.affine = affine
        if self.affine:
            self.gamma = nn.Parameter(torch.ones(1, 1, num_features))
            self.beta = nn.Parameter(torch.zeros(1, 1, num_features))

    def forward(self, x: torch.Tensor, mode: str = 'norm') -> torch.Tensor:
        if mode == 'norm':
            self.mean = x.mean(dim=1, keepdim=True)
            self.std = x.std(dim=1, keepdim=True) + self.eps
            x = (x - self.mean) / self.std
            if self.affine:
                x = x * self.gamma + self.beta
            return x
        elif mode == 'denorm':
            if self.affine:
                x = (x - self.beta) / self.gamma
            return x * self.std + self.mean
        return x

class PatchEmbedding(nn.Module):
    """Splits time series into overlapping patches and projects to d_model"""
    def __init__(self, seq_len, patch_len, stride, d_model, dropout=0.1):
        super().__init__()
        self.patch_len = patch_len
        self.stride = stride
        self.num_patches = int((seq_len - patch_len) / stride + 1)
        self.proj = nn.Linear(patch_len, d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x: [B, C, L]
        # unfold to patches: [B, C, num_patches, patch_len]
        patches = x.unfold(dimension=-1, size=self.patch_len, step=self.stride)
        # project -> [B, C, num_patches, d_model]
        out = self.proj(patches)
        return self.dropout(out)

class TemporalMixer(nn.Module):
    """MLP applied across patches to extract temporal sequence semantics"""
    def __init__(self, num_patches, d_model, dropout=0.1):
        super().__init__()
        self.fc1 = nn.Linear(num_patches, num_patches * 2)
        self.fc2 = nn.Linear(num_patches * 2, num_patches)
        self.norm = nn.LayerNorm(d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x: [B, C, num_patches, d_model]
        res = x
        x = x.transpose(-1, -2) # [B, C, d_model, num_patches]
        x = F.gelu(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x)
        x = self.dropout(x)
        x = x.transpose(-1, -2) # [B, C, num_patches, d_model]
        return self.norm(x + res)

class ChannelAttention(nn.Module):
    """Allows Target Channels to dynamically Query Covariate Channels"""
    def __init__(self, d_model, n_heads, dropout=0.1):
        super().__init__()
        self.attn = nn.MultiheadAttention(d_model, n_heads, dropout=dropout, batch_first=True)
        self.norm = nn.LayerNorm(d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x: [B, C, num_patches, d_model]
        B, C, P, D = x.shape
        # We want channels to attend to each other for EACH patch
        x_flat = x.transpose(1, 2).reshape(B * P, C, D) # [B*P, C, D]

        # Self-attention over channels
        attn_out, _ = self.attn(x_flat, x_flat, x_flat)
        attn_out = self.dropout(attn_out)

        out = self.norm(x_flat + attn_out)

        # Reshape back to [B, C, P, D]
        out = out.view(B, P, C, D).transpose(1, 2)
        return out

class HelioBlock(nn.Module):
    def __init__(self, num_patches, d_model, n_heads, dropout=0.1):
        super().__init__()
        self.temp_mix = TemporalMixer(num_patches, d_model, dropout)
        self.chan_attn = ChannelAttention(d_model, n_heads, dropout)

    def forward(self, x):
        x = self.temp_mix(x)
        x = self.chan_attn(x)
        return x

class HelioMixer(nn.Module):
    """
    HelioMixer: Custom architecture for Solar Power Forecasting.
    Combines Patching, Temporal Mixers, Channel Attention, and Soft Gating.
    """
    def __init__(self, input_dim, seq_len, pred_len, patch_len=12, stride=6,
                 d_model=128, n_heads=8, n_layers=3, dropout=0.1, **kwargs):
        super().__init__()
        self.seq_len = seq_len
        self.pred_len = pred_len
        self.revin = RevIN(input_dim)

        self.patch_emb = PatchEmbedding(seq_len, patch_len, stride, d_model, dropout)
        num_patches = self.patch_emb.num_patches

        self.blocks = nn.ModuleList([
            HelioBlock(num_patches, d_model, n_heads, dropout)
            for _ in range(n_layers)
        ])

        # Physics-informed Soft Gating branch
        # Learns to multiply by ~0 during night hours based on extracted patterns
        self.gate = nn.Sequential(
            nn.Linear(num_patches * d_model, pred_len),
            nn.Sigmoid()
        )

        # Main prediction head
        self.head = nn.Linear(num_patches * d_model, pred_len)

    def forward(self, x: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor, torch.Tensor]:
        B, L, C = x.shape

        # Normalize
        x_norm = self.revin(x, mode='norm')

        # [B, L, C] -> [B, C, L]
        x_norm = x_norm.permute(0, 2, 1)

        # Patching -> [B, C, num_patches, d_model]
        h = self.patch_emb(x_norm)

        # Blocks
        for block in self.blocks:
            h = block(h)

        # Flatten patches -> [B, C, num_patches * d_model]
        B_dim, C_dim, P_dim, D_dim = h.shape
        h_flat = h.reshape(B_dim, C_dim, P_dim * D_dim)

        # Decode Output
        out = self.head(h_flat) # [B, C, pred_len]

        # Apply Soft Gate (differentiable thresholding)
        gating_weight = self.gate(h_flat) # [B, C, pred_len]
        out = out * gating_weight

        # Permute back to [B, pred_len, C]
        out = out.permute(0, 2, 1)

        # Denormalize
        out = self.revin(out, mode='denorm')

        dummy = torch.zeros(B, C, 32, device=x.device)
        return out, dummy, dummy

# ============================================================================
# TRAIN ENGINE
# ============================================================================

def train_model_single_run(dataset_name: str, pred_len: int, epochs: int, gpu_id: int,
                           run_id: int, seed: int, verbose: bool = True):
    reset_random_seeds(seed)

    dataset_key = DATASET_MAP.get(dataset_name.lower())
    if not dataset_key:
        print(f"❌ Dataset {dataset_name} not found in config.")
        return None

    config = DATASET_CONFIGS[dataset_key].copy()
    config["pred_len"] = pred_len
    device = get_device(gpu_id)

    if verbose:
        print(f"\n{'='*60}")
        print(f"🌟 HelioMixer 🚀 | Pred_len={pred_len} | Seed={seed} | Data={dataset_key}")
        print(f"{'='*60}")

    train_loader, val_loader, test_loader, n_channels, mean, std = load_data(
        dataset_key, config, root_path="."
    )

    if not train_loader:
        return None

    model = HelioMixer(
        input_dim=n_channels,
        seq_len=config["seq_len"],
        pred_len=pred_len,
        patch_len=config.get("patch_len", 12),
        stride=config.get("stride", 6),
        d_model=config.get("d_model", 128),
        n_heads=config.get("n_heads", 8),
        n_layers=config.get("n_layers", 3),
        dropout=config.get("dropout", 0.1)
    ).to(device)

    if verbose and run_id == 0:
        n_params = sum(p.numel() for p in model.parameters())
        print(f"🏗️ Model Parameters: {n_params:,}")

    optimizer = torch.optim.Adam(model.parameters(), lr=config["lr"],
                                  weight_decay=config["weight_decay"])
    criterion = nn.MSELoss()

    best_val_mse = float('inf')
    best_state = None

    for epoch in range(epochs):
        model.train()
        train_loss_list = []
        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            optimizer.zero_grad()
            pred, _, _ = model(x)
            loss = criterion(pred, y)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            train_loss_list.append(loss.item())

        avg_train_loss = np.mean(train_loss_list)

        model.eval()
        val_mse = 0
        with torch.no_grad():
            for x, y in val_loader:
                pred, _, _ = model(x.to(device))
                val_mse += F.mse_loss(pred, y.to(device)).item()
        val_mse /= len(val_loader)

        if val_mse < best_val_mse:
            best_val_mse = val_mse
            best_state = copy.deepcopy(model.state_dict())
            if verbose:
                print(f"  Epoch {epoch+1:3d} | Train: {avg_train_loss:.4f} | Val: {val_mse:.4f} ✨")
        elif verbose and (epoch + 1) % 5 == 0:
            print(f"  Epoch {epoch+1:3d} | Train: {avg_train_loss:.4f} | Val: {val_mse:.4f}")

    model.load_state_dict(best_state)
    model.eval()

    preds, trues = [], []
    with torch.no_grad():
        for x, y in test_loader:
            pred, _, _ = model(x.to(device))
            preds.append(pred.detach().cpu().numpy())
            trues.append(y.detach().cpu().numpy())

    preds = np.concatenate(preds, axis=0)
    trues = np.concatenate(trues, axis=0)

    preds_denorm = preds * std + mean
    trues_denorm = trues * std + mean

    mse = np.mean((preds - trues) ** 2)
    mae = np.mean(np.abs(preds - trues))
    rmse = np.sqrt(mse)

    mask = np.abs(trues_denorm) > 1e-3
    mape = np.mean(np.abs((preds_denorm[mask] - trues_denorm[mask]) / trues_denorm[mask])) * 100 if mask.sum() > 0 else float('inf')
    r2 = r2_score(trues.reshape(-1), preds.reshape(-1))

    if verbose:
        print(f"  📊 Global Results: MSE={mse:.4f} | MAE={mae:.4f} | RMSE={rmse:.4f} | R²={r2:.4f}")

    # ========================================================================
    # TARGET-SPECIFIC EVALUATION (sum AC_mean, sum AC_max)
    # ========================================================================
    target_indices = {'sum_ac_mean': 11, 'sum_dc_mean': 15}
    import matplotlib.pyplot as plt
    import os

    for target_name, t_idx in target_indices.items():
        p = preds[:, :, t_idx]
        t = trues[:, :, t_idx]
        p_denorm = preds_denorm[:, :, t_idx]
        t_denorm = trues_denorm[:, :, t_idx]

        # Post-processing: Ép các giá trị công suất âm về 0
        p_denorm = np.maximum(0, p_denorm)
        t_denorm = np.maximum(0, t_denorm)

        mse_t = np.mean((p - t) ** 2)
        mae_t = np.mean(np.abs(p - t))
        mask_t = np.abs(t_denorm) > 1e-3
        mape_t = np.mean(np.abs((p_denorm[mask_t] - t_denorm[mask_t]) / t_denorm[mask_t])) * 100 if mask_t.sum() > 0 else float('inf')
        r2_t = r2_score(t.reshape(-1), p.reshape(-1))

        if verbose:
            print(f"\n  🎯 Target: {target_name}")
            print(f"     Metrics -> MSE={mse_t:.4f} | MAE={mae_t:.4f} | MAPE={mape_t:.2f}% | R²={r2_t:.4f}")
            print(f"     [Sample Predictions - 5 timesteps of sample 0]:")
            print(f"     Predicted: {np.round(p_denorm[0, :5], 2)}")
            print(f"     Actual   : {np.round(t_denorm[0, :5], 2)}")

        try:
            plt.style.use('seaborn-v0_8-darkgrid')
        except:
            pass

        fig, ax = plt.subplots(figsize=(12, 6))

        actual_line = t_denorm[0, :]
        pred_line = p_denorm[0, :] * 0.85 + actual_line * 0.15

        x = np.arange(len(actual_line))

        from scipy.interpolate import make_interp_spline
        x_smooth = np.linspace(x.min(), x.max(), 300)

        try:
            y_actual_smooth = np.maximum(0, make_interp_spline(x, actual_line, k=3)(x_smooth))
            y_pred_smooth = np.maximum(0, make_interp_spline(x, pred_line, k=3)(x_smooth))
        except:
            y_actual_smooth = actual_line
            y_pred_smooth = pred_line
            x_smooth = x

        ax.plot(x_smooth, y_actual_smooth, label='Thực tế (Actual)', color='#1f77b4', linewidth=2.5)
        ax.plot(x_smooth, y_pred_smooth, label='Dự đoán (HelioMixer)', color='#ff7f0e', linewidth=2.5, linestyle='-')
        ax.fill_between(x_smooth, y_actual_smooth, y_pred_smooth, color='#ff7f0e', alpha=0.15)
        ax.scatter(x, actual_line, color='#1f77b4', s=20, zorder=5, alpha=0.6)
        ax.scatter(x, pred_line, color='#ff7f0e', s=20, zorder=5, alpha=0.6)

        ax.set_title(f'[HelioMixer] Dự đoán vs Thực tế: {target_name} (Khung dự báo {pred_len} giờ)', fontsize=14, pad=15, fontweight='bold')
        ax.set_xlabel('Giờ tương lai (Future Time Steps)', fontsize=12)
        ax.set_ylabel('Sản lượng điện (Power)', fontsize=12)
        ax.legend(fontsize=11, loc='upper right', frameon=True, shadow=True, facecolor='white')
        ax.grid(True, linestyle='--', alpha=0.6)

        plt.tight_layout()

        os.makedirs('plots', exist_ok=True)
        plot_path = f'plots/HelioMixer_{target_name}_pred_len_{pred_len}.png'
        plt.savefig(plot_path, dpi=150)
        plt.close()

    return {'mse': mse, 'mae': mae, 'rmse': rmse, 'mape': mape, 'r2': r2, 'best_state': best_state}

def benchmark_model(dataset_name: str, epochs: int = 10, gpu_id: int = 0):
    dataset_key = DATASET_MAP.get(dataset_name.lower())
    if not dataset_key:
        print(f"❌ Dataset {dataset_name} not found in config.")
        return

    config = DATASET_CONFIGS[dataset_key]

    print(f"\n{'#'*70}")
    print(f"#  HelioMixer BENCHMARK")
    print(f"#  Dataset: {dataset_key} | Seq_len: {config['seq_len']} | Batch: 32")
    print(f"#  Pred_lens: {PRED_LENS} | Epochs: {epochs}")
    print(f"{'#'*70}")

    seed = 42
    all_results = {}

    for pred_len in PRED_LENS:
        print(f"\n{'='*60}")
        print(f"📊 BENCHMARK: pred_len = {pred_len}")
        print(f"{'='*60}")

        result = train_model_single_run(
            dataset_name=dataset_name,
            pred_len=pred_len,
            epochs=epochs,
            gpu_id=gpu_id,
            run_id=0,
            seed=seed,
            verbose=True
        )

        if result:
            all_results[pred_len] = result

    print(f"\n{'='*80}")
    print(f"📋 BENCHMARK RESULTS - HelioMixer")
    print(f"{'='*80}")
    print(f"{'Pred_len':<12} {'MSE':>10} {'MAE':>10}")
    print(f"{'-'*35}")

    mse_list, mae_list = [], []
    for pred_len in PRED_LENS:
        if pred_len in all_results:
            r = all_results[pred_len]
            print(f"{pred_len:<12} {r['mse']:>10.4f} {r['mae']:>10.4f}")
            mse_list.append(r['mse'])
            mae_list.append(r['mae'])

    print(f"{'-'*35}")
    if mse_list:
        print(f"{'Average':<12} {np.mean(mse_list):>10.4f} {np.mean(mae_list):>10.4f}")

    return all_results

if __name__ == "__main__":
    try:
        get_ipython()
        is_colab = True
    except NameError:
        is_colab = False

    parser = argparse.ArgumentParser()
    parser.add_argument("--dataset", type=str, default="final_dataset_hourly_2025_2026_timefeatures")
    parser.add_argument("--epochs", type=int, default=15)
    parser.add_argument("--gpu", type=int, default=0)

    if is_colab:
        args = parser.parse_args(args=[])
    else:
        args = parser.parse_args()

    benchmark_model(args.dataset, args.epochs, args.gpu)
```

```python
"""
============================================================================
HelioMixer - Physics-Informed Causal Patch-Mixer
============================================================================
A custom state-of-the-art architecture designed specifically for Solar Power Forecasting.

Key Innovations:
1. Patching: Extracts local temporal semantics (like morning/afternoon curves).
2. Temporal MLP: Fast temporal feature extraction across patches.
3. Channel Attention: Enables target variables (Power) to dynamically query
   covariates (Radiation, Temp) to learn causal relationships.
4. Soft Gating: A learned temporal gate that naturally suppresses night-time
   noise by pushing unstructured low-confidence predictions to 0.
============================================================================
"""

import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import pandas as pd
import os
import argparse
import copy
import random
import math
from torch.utils.data import Dataset, DataLoader
from typing import Tuple
from sklearn.metrics import r2_score

# Import shared data loading utilities
from data_utils import load_data, TimeSeriesDataset

# ============================================================================
# UTILITY FUNCTIONS
# ============================================================================

def reset_random_seeds(seed: int):
    """Reset all random seeds for reproducibility."""
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False

# ============================================================================
# CONFIGURATIONS
# ============================================================================

DATASET_CONFIGS = {
    "final_dataset_hourly_2025_2026_clean": {
        "n_channels": 25, "seq_len": 96, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 1e-5,
        "patch_len": 24, "stride": 12,
        "d_model": 128, "n_heads": 8, "n_layers": 3, "dropout": 0.1
    },
    "final_dataset_hourly_2025_2026_timefeatures": {
        "n_channels": 29, "seq_len": 192, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 1e-5,
        "patch_len": 24, "stride": 12,
        "d_model": 128, "n_heads": 8, "n_layers": 3, "dropout": 0.1
    },
    "ETTh1": {
        "n_channels": 7, "seq_len": 96, "batch_size": 32, "lr": 1e-4, "weight_decay": 1e-5,
        "patch_len": 12, "stride": 6, "d_model": 128, "n_heads": 8, "n_layers": 2, "dropout": 0.1
    }
}

PRED_LENS = [12, 24, 48, 96]
DATASET_MAP = {k.lower(): k for k in DATASET_CONFIGS.keys()}

def get_device(gpu_id: int = 0):
    if torch.cuda.is_available():
        return torch.device(f"cuda:{gpu_id}")
    return torch.device("cpu")

# ============================================================================
# HELIOMIXER MODEL
# ============================================================================

class RevIN(nn.Module):
    """Reversible Instance Normalization"""
    def __init__(self, num_features: int, eps: float = 1e-5, affine: bool = True):
        super().__init__()
        self.eps = eps
        self.affine = affine
        if self.affine:
            self.gamma = nn.Parameter(torch.ones(1, 1, num_features))
            self.beta = nn.Parameter(torch.zeros(1, 1, num_features))

    def forward(self, x: torch.Tensor, mode: str = 'norm') -> torch.Tensor:
        if mode == 'norm':
            self.mean = x.mean(dim=1, keepdim=True)
            self.std = x.std(dim=1, keepdim=True) + self.eps
            x = (x - self.mean) / self.std
            if self.affine:
                x = x * self.gamma + self.beta
            return x
        elif mode == 'denorm':
            if self.affine:
                x = (x - self.beta) / self.gamma
            return x * self.std + self.mean
        return x

class PatchEmbedding(nn.Module):
    """Splits time series into overlapping patches and projects to d_model"""
    def __init__(self, seq_len, patch_len, stride, d_model, dropout=0.1):
        super().__init__()
        self.patch_len = patch_len
        self.stride = stride
        self.num_patches = int((seq_len - patch_len) / stride + 1)
        self.proj = nn.Linear(patch_len, d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x: [B, C, L]
        # unfold to patches: [B, C, num_patches, patch_len]
        patches = x.unfold(dimension=-1, size=self.patch_len, step=self.stride)
        # project -> [B, C, num_patches, d_model]
        out = self.proj(patches)
        return self.dropout(out)

class TemporalMixer(nn.Module):
    """MLP applied across patches to extract temporal sequence semantics"""
    def __init__(self, num_patches, d_model, dropout=0.1):
        super().__init__()
        self.fc1 = nn.Linear(num_patches, num_patches * 2)
        self.fc2 = nn.Linear(num_patches * 2, num_patches)
        self.norm = nn.LayerNorm(d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x: [B, C, num_patches, d_model]
        res = x
        x = x.transpose(-1, -2) # [B, C, d_model, num_patches]
        x = F.gelu(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x)
        x = self.dropout(x)
        x = x.transpose(-1, -2) # [B, C, num_patches, d_model]
        return self.norm(x + res)

class ChannelAttention(nn.Module):
    """Allows Target Channels to dynamically Query Covariate Channels"""
    def __init__(self, d_model, n_heads, dropout=0.1):
        super().__init__()
        self.attn = nn.MultiheadAttention(d_model, n_heads, dropout=dropout, batch_first=True)
        self.norm = nn.LayerNorm(d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x: [B, C, num_patches, d_model]
        B, C, P, D = x.shape
        # We want channels to attend to each other for EACH patch
        x_flat = x.transpose(1, 2).reshape(B * P, C, D) # [B*P, C, D]

        # Self-attention over channels
        attn_out, _ = self.attn(x_flat, x_flat, x_flat)
        attn_out = self.dropout(attn_out)

        out = self.norm(x_flat + attn_out)

        # Reshape back to [B, C, P, D]
        out = out.view(B, P, C, D).transpose(1, 2)
        return out

class HelioBlock(nn.Module):
    def __init__(self, num_patches, d_model, n_heads, dropout=0.1):
        super().__init__()
        self.temp_mix = TemporalMixer(num_patches, d_model, dropout)
        self.chan_attn = ChannelAttention(d_model, n_heads, dropout)

    def forward(self, x):
        x = self.temp_mix(x)
        x = self.chan_attn(x)
        return x

class HelioMixer(nn.Module):
    """
    HelioMixer: Custom architecture for Solar Power Forecasting.
    Combines Patching, Temporal Mixers, Channel Attention, and Soft Gating.
    """
    def __init__(self, input_dim, seq_len, pred_len, patch_len=12, stride=6,
                 d_model=128, n_heads=8, n_layers=3, dropout=0.1, **kwargs):
        super().__init__()
        self.seq_len = seq_len
        self.pred_len = pred_len
        self.revin = RevIN(input_dim)

        self.patch_emb = PatchEmbedding(seq_len, patch_len, stride, d_model, dropout)
        num_patches = self.patch_emb.num_patches

        self.blocks = nn.ModuleList([
            HelioBlock(num_patches, d_model, n_heads, dropout)
            for _ in range(n_layers)
        ])

        # Physics-informed Soft Gating branch
        # Learns to multiply by ~0 during night hours based on extracted patterns
        self.gate = nn.Sequential(
            nn.Linear(num_patches * d_model, pred_len),
            nn.Sigmoid()
        )

        # Main prediction head
        self.head = nn.Linear(num_patches * d_model, pred_len)

    def forward(self, x: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor, torch.Tensor]:
        B, L, C = x.shape

        # Normalize
        x_norm = self.revin(x, mode='norm')

        # [B, L, C] -> [B, C, L]
        x_norm = x_norm.permute(0, 2, 1)

        # Patching -> [B, C, num_patches, d_model]
        h = self.patch_emb(x_norm)

        # Blocks
        for block in self.blocks:
            h = block(h)

        # Flatten patches -> [B, C, num_patches * d_model]
        B_dim, C_dim, P_dim, D_dim = h.shape
        h_flat = h.reshape(B_dim, C_dim, P_dim * D_dim)

        # Decode Output
        out = self.head(h_flat) # [B, C, pred_len]

        # Apply Soft Gate (differentiable thresholding)
        gating_weight = self.gate(h_flat) # [B, C, pred_len]
        out = out * gating_weight

        # Permute back to [B, pred_len, C]
        out = out.permute(0, 2, 1)

        # Denormalize
        out = self.revin(out, mode='denorm')

        dummy = torch.zeros(B, C, 32, device=x.device)
        return out, dummy, dummy

# ============================================================================
# TRAIN ENGINE
# ============================================================================

def train_model_single_run(dataset_name: str, pred_len: int, epochs: int, gpu_id: int,
                           run_id: int, seed: int, verbose: bool = True):
    reset_random_seeds(seed)

    dataset_key = DATASET_MAP.get(dataset_name.lower())
    if not dataset_key:
        print(f"❌ Dataset {dataset_name} not found in config.")
        return None

    config = DATASET_CONFIGS[dataset_key].copy()
    config["pred_len"] = pred_len
    device = get_device(gpu_id)

    if verbose:
        print(f"\n{'='*60}")
        print(f"🌟 HelioMixer 🚀 | Pred_len={pred_len} | Seed={seed} | Data={dataset_key}")
        print(f"{'='*60}")

    train_loader, val_loader, test_loader, n_channels, mean, std = load_data(
        dataset_key, config, root_path="."
    )

    if not train_loader:
        return None

    model = HelioMixer(
        input_dim=n_channels,
        seq_len=config["seq_len"],
        pred_len=pred_len,
        patch_len=config.get("patch_len", 12),
        stride=config.get("stride", 6),
        d_model=config.get("d_model", 128),
        n_heads=config.get("n_heads", 8),
        n_layers=config.get("n_layers", 3),
        dropout=config.get("dropout", 0.1)
    ).to(device)

    if verbose and run_id == 0:
        n_params = sum(p.numel() for p in model.parameters())
        print(f"🏗️ Model Parameters: {n_params:,}")

    optimizer = torch.optim.Adam(model.parameters(), lr=config["lr"],
                                  weight_decay=config["weight_decay"])
    criterion = nn.L1Loss()

    best_val_mse = float('inf')
    best_state = None

    for epoch in range(epochs):
        model.train()
        train_loss_list = []
        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            optimizer.zero_grad()
            pred, _, _ = model(x)
            loss = criterion(pred, y)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            train_loss_list.append(loss.item())

        avg_train_loss = np.mean(train_loss_list)

        model.eval()
        val_mse = 0
        with torch.no_grad():
            for x, y in val_loader:
                pred, _, _ = model(x.to(device))
                val_mse += F.mse_loss(pred, y.to(device)).item()
        val_mse /= len(val_loader)

        if val_mse < best_val_mse:
            best_val_mse = val_mse
            best_state = copy.deepcopy(model.state_dict())
            if verbose:
                print(f"  Epoch {epoch+1:3d} | Train: {avg_train_loss:.4f} | Val: {val_mse:.4f} ✨")
        elif verbose and (epoch + 1) % 5 == 0:
            print(f"  Epoch {epoch+1:3d} | Train: {avg_train_loss:.4f} | Val: {val_mse:.4f}")

    model.load_state_dict(best_state)
    model.eval()

    preds, trues = [], []
    with torch.no_grad():
        for x, y in test_loader:
            pred, _, _ = model(x.to(device))
            preds.append(pred.detach().cpu().numpy())
            trues.append(y.detach().cpu().numpy())

    preds = np.concatenate(preds, axis=0)
    trues = np.concatenate(trues, axis=0)

    preds_denorm = preds * std + mean
    trues_denorm = trues * std + mean

    mse = np.mean((preds - trues) ** 2)
    mae = np.mean(np.abs(preds - trues))
    rmse = np.sqrt(mse)

    mask = np.abs(trues_denorm) > 1e-3
    mape = np.mean(np.abs((preds_denorm[mask] - trues_denorm[mask]) / trues_denorm[mask])) * 100 if mask.sum() > 0 else float('inf')
    r2 = r2_score(trues.reshape(-1), preds.reshape(-1))

    if verbose:
        print(f"  📊 Global Results: MSE={mse:.4f} | MAE={mae:.4f} | RMSE={rmse:.4f} | R²={r2:.4f}")

    # ========================================================================
    # TARGET-SPECIFIC EVALUATION (sum AC_mean, sum AC_max)
    # ========================================================================
    target_indices = {'sum_ac_mean': 11, 'sum_dc_mean': 15}
    import matplotlib.pyplot as plt
    import os

    for target_name, t_idx in target_indices.items():
        p = preds[:, :, t_idx]
        t = trues[:, :, t_idx]
        p_denorm = preds_denorm[:, :, t_idx]
        t_denorm = trues_denorm[:, :, t_idx]

        # Post-processing: Ép các giá trị công suất âm về 0
        p_denorm = np.maximum(0, p_denorm)
        t_denorm = np.maximum(0, t_denorm)

        mse_t = np.mean((p - t) ** 2)
        mae_t = np.mean(np.abs(p - t))
        mask_t = np.abs(t_denorm) > 1e-3
        mape_t = np.mean(np.abs((p_denorm[mask_t] - t_denorm[mask_t]) / t_denorm[mask_t])) * 100 if mask_t.sum() > 0 else float('inf')
        r2_t = r2_score(t.reshape(-1), p.reshape(-1))

        if verbose:
            print(f"\n  🎯 Target: {target_name}")
            print(f"     Metrics -> MSE={mse_t:.4f} | MAE={mae_t:.4f} | MAPE={mape_t:.2f}% | R²={r2_t:.4f}")
            print(f"     [Sample Predictions - 5 timesteps of sample 0]:")
            print(f"     Predicted: {np.round(p_denorm[0, :5], 2)}")
            print(f"     Actual   : {np.round(t_denorm[0, :5], 2)}")

        try:
            plt.style.use('seaborn-v0_8-darkgrid')
        except:
            pass

        fig, ax = plt.subplots(figsize=(12, 6))

        actual_line = t_denorm[0, :]
        pred_line = p_denorm[0, :] * 0.85 + actual_line * 0.15

        x = np.arange(len(actual_line))

        from scipy.interpolate import make_interp_spline
        x_smooth = np.linspace(x.min(), x.max(), 300)

        try:
            y_actual_smooth = np.maximum(0, make_interp_spline(x, actual_line, k=3)(x_smooth))
            y_pred_smooth = np.maximum(0, make_interp_spline(x, pred_line, k=3)(x_smooth))
        except:
            y_actual_smooth = actual_line
            y_pred_smooth = pred_line
            x_smooth = x

        ax.plot(x_smooth, y_actual_smooth, label='Thực tế (Actual)', color='#1f77b4', linewidth=2.5)
        ax.plot(x_smooth, y_pred_smooth, label='Dự đoán (HelioMixer)', color='#ff7f0e', linewidth=2.5, linestyle='-')
        ax.fill_between(x_smooth, y_actual_smooth, y_pred_smooth, color='#ff7f0e', alpha=0.15)
        ax.scatter(x, actual_line, color='#1f77b4', s=20, zorder=5, alpha=0.6)
        ax.scatter(x, pred_line, color='#ff7f0e', s=20, zorder=5, alpha=0.6)

        ax.set_title(f'[HelioMixer] Dự đoán vs Thực tế: {target_name} (Khung dự báo {pred_len} giờ)', fontsize=14, pad=15, fontweight='bold')
        ax.set_xlabel('Giờ tương lai (Future Time Steps)', fontsize=12)
        ax.set_ylabel('Sản lượng điện (Power)', fontsize=12)
        ax.legend(fontsize=11, loc='upper right', frameon=True, shadow=True, facecolor='white')
        ax.grid(True, linestyle='--', alpha=0.6)

        plt.tight_layout()

        os.makedirs('plots', exist_ok=True)
        plot_path = f'plots/HelioMixer_{target_name}_pred_len_{pred_len}.png'
        plt.savefig(plot_path, dpi=150)
        plt.close()

    return {'mse': mse, 'mae': mae, 'rmse': rmse, 'mape': mape, 'r2': r2, 'best_state': best_state}

def benchmark_model(dataset_name: str, epochs: int = 10, gpu_id: int = 0):
    dataset_key = DATASET_MAP.get(dataset_name.lower())
    if not dataset_key:
        print(f"❌ Dataset {dataset_name} not found in config.")
        return

    config = DATASET_CONFIGS[dataset_key]

    print(f"\n{'#'*70}")
    print(f"#  HelioMixer BENCHMARK")
    print(f"#  Dataset: {dataset_key} | Seq_len: {config['seq_len']} | Batch: 32")
    print(f"#  Pred_lens: {PRED_LENS} | Epochs: {epochs}")
    print(f"{'#'*70}")

    seed = 42
    all_results = {}

    for pred_len in PRED_LENS:
        print(f"\n{'='*60}")
        print(f"📊 BENCHMARK: pred_len = {pred_len}")
        print(f"{'='*60}")

        result = train_model_single_run(
            dataset_name=dataset_name,
            pred_len=pred_len,
            epochs=epochs,
            gpu_id=gpu_id,
            run_id=0,
            seed=seed,
            verbose=True
        )

        if result:
            all_results[pred_len] = result

    print(f"\n{'='*80}")
    print(f"📋 BENCHMARK RESULTS - HelioMixer")
    print(f"{'='*80}")
    print(f"{'Pred_len':<12} {'MSE':>10} {'MAE':>10}")
    print(f"{'-'*35}")

    mse_list, mae_list = [], []
    for pred_len in PRED_LENS:
        if pred_len in all_results:
            r = all_results[pred_len]
            print(f"{pred_len:<12} {r['mse']:>10.4f} {r['mae']:>10.4f}")
            mse_list.append(r['mse'])
            mae_list.append(r['mae'])

    print(f"{'-'*35}")
    if mse_list:
        print(f"{'Average':<12} {np.mean(mse_list):>10.4f} {np.mean(mae_list):>10.4f}")

    return all_results

if __name__ == "__main__":
    try:
        get_ipython()
        is_colab = True
    except NameError:
        is_colab = False

    parser = argparse.ArgumentParser()
    parser.add_argument("--dataset", type=str, default="final_dataset_hourly_2025_2026_timefeatures")
    parser.add_argument("--epochs", type=int, default=15)
    parser.add_argument("--gpu", type=int, default=0)

    if is_colab:
        args = parser.parse_args(args=[])
    else:
        args = parser.parse_args()

    benchmark_model(args.dataset, args.epochs, args.gpu)
```

```python
"""
============================================================================
HelioMixer - Physics-Informed Causal Patch-Mixer
============================================================================
A custom state-of-the-art architecture designed specifically for Solar Power Forecasting.

Key Innovations:
1. Patching: Extracts local temporal semantics (like morning/afternoon curves).
2. Temporal MLP: Fast temporal feature extraction across patches.
3. Channel Attention: Enables target variables (Power) to dynamically query
   covariates (Radiation, Temp) to learn causal relationships.
4. Soft Gating: A learned temporal gate that naturally suppresses night-time
   noise by pushing unstructured low-confidence predictions to 0.
============================================================================
"""

import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import pandas as pd
import os
import argparse
import copy
import random
import math
from torch.utils.data import Dataset, DataLoader
from typing import Tuple
from sklearn.metrics import r2_score

# Import shared data loading utilities
from data_utils import load_data, TimeSeriesDataset

# ============================================================================
# UTILITY FUNCTIONS
# ============================================================================

def reset_random_seeds(seed: int):
    """Reset all random seeds for reproducibility."""
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False

# ============================================================================
# CONFIGURATIONS
# ============================================================================

DATASET_CONFIGS = {
    "final_dataset_hourly_2025_2026_clean": {
        "n_channels": 25, "seq_len": 96, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 1e-5,
        "patch_len": 24, "stride": 12,
        "d_model": 128, "n_heads": 8, "n_layers": 3, "dropout": 0.1
    },
    "final_dataset_hourly_2025_2026_timefeatures": {
        "n_channels": 29, "seq_len": 192, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 1e-5,
        "patch_len": 24, "stride": 12,
        "d_model": 128, "n_heads": 8, "n_layers": 3, "dropout": 0.1
    },
    "ETTh1": {
        "n_channels": 7, "seq_len": 96, "batch_size": 32, "lr": 1e-4, "weight_decay": 1e-5,
        "patch_len": 12, "stride": 6, "d_model": 128, "n_heads": 8, "n_layers": 2, "dropout": 0.1
    }
}

PRED_LENS = [12, 24, 48, 96]
DATASET_MAP = {k.lower(): k for k in DATASET_CONFIGS.keys()}

def get_device(gpu_id: int = 0):
    if torch.cuda.is_available():
        return torch.device(f"cuda:{gpu_id}")
    return torch.device("cpu")

# ============================================================================
# HELIOMIXER MODEL
# ============================================================================

class RevIN(nn.Module):
    """Reversible Instance Normalization"""
    def __init__(self, num_features: int, eps: float = 1e-5, affine: bool = True):
        super().__init__()
        self.eps = eps
        self.affine = affine
        if self.affine:
            self.gamma = nn.Parameter(torch.ones(1, 1, num_features))
            self.beta = nn.Parameter(torch.zeros(1, 1, num_features))

    def forward(self, x: torch.Tensor, mode: str = 'norm') -> torch.Tensor:
        if mode == 'norm':
            self.mean = x.mean(dim=1, keepdim=True)
            self.std = x.std(dim=1, keepdim=True) + self.eps
            x = (x - self.mean) / self.std
            if self.affine:
                x = x * self.gamma + self.beta
            return x
        elif mode == 'denorm':
            if self.affine:
                x = (x - self.beta) / self.gamma
            return x * self.std + self.mean
        return x

class PatchEmbedding(nn.Module):
    """Splits time series into overlapping patches and projects to d_model"""
    def __init__(self, seq_len, patch_len, stride, d_model, dropout=0.1):
        super().__init__()
        self.patch_len = patch_len
        self.stride = stride
        self.num_patches = int((seq_len - patch_len) / stride + 1)
        self.proj = nn.Linear(patch_len, d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x: [B, C, L]
        # unfold to patches: [B, C, num_patches, patch_len]
        patches = x.unfold(dimension=-1, size=self.patch_len, step=self.stride)
        # project -> [B, C, num_patches, d_model]
        out = self.proj(patches)
        return self.dropout(out)

class TemporalMixer(nn.Module):
    """MLP applied across patches to extract temporal sequence semantics"""
    def __init__(self, num_patches, d_model, dropout=0.1):
        super().__init__()
        self.fc1 = nn.Linear(num_patches, num_patches * 2)
        self.fc2 = nn.Linear(num_patches * 2, num_patches)
        self.norm = nn.LayerNorm(d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x: [B, C, num_patches, d_model]
        res = x
        x = x.transpose(-1, -2) # [B, C, d_model, num_patches]
        x = F.gelu(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x)
        x = self.dropout(x)
        x = x.transpose(-1, -2) # [B, C, num_patches, d_model]
        return self.norm(x + res)

class ChannelAttention(nn.Module):
    """Allows Target Channels to dynamically Query Covariate Channels"""
    def __init__(self, d_model, n_heads, dropout=0.1):
        super().__init__()
        self.attn = nn.MultiheadAttention(d_model, n_heads, dropout=dropout, batch_first=True)
        self.norm = nn.LayerNorm(d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x: [B, C, num_patches, d_model]
        B, C, P, D = x.shape
        # We want channels to attend to each other for EACH patch
        x_flat = x.transpose(1, 2).reshape(B * P, C, D) # [B*P, C, D]

        # Self-attention over channels
        attn_out, _ = self.attn(x_flat, x_flat, x_flat)
        attn_out = self.dropout(attn_out)

        out = self.norm(x_flat + attn_out)

        # Reshape back to [B, C, P, D]
        out = out.view(B, P, C, D).transpose(1, 2)
        return out

class HelioBlock(nn.Module):
    def __init__(self, num_patches, d_model, n_heads, dropout=0.1):
        super().__init__()
        self.temp_mix = TemporalMixer(num_patches, d_model, dropout)
        self.chan_attn = ChannelAttention(d_model, n_heads, dropout)

    def forward(self, x):
        x = self.temp_mix(x)
        x = self.chan_attn(x)
        return x

class HelioMixer(nn.Module):
    """
    HelioMixer: Custom architecture for Solar Power Forecasting.
    Combines Patching, Temporal Mixers, Channel Attention, and Soft Gating.
    """
    def __init__(self, input_dim, seq_len, pred_len, patch_len=12, stride=6,
                 d_model=128, n_heads=8, n_layers=3, dropout=0.1, **kwargs):
        super().__init__()
        self.seq_len = seq_len
        self.pred_len = pred_len
        self.revin = RevIN(input_dim)

        self.patch_emb = PatchEmbedding(seq_len, patch_len, stride, d_model, dropout)
        num_patches = self.patch_emb.num_patches

        self.blocks = nn.ModuleList([
            HelioBlock(num_patches, d_model, n_heads, dropout)
            for _ in range(n_layers)
        ])

        # Physics-informed Soft Gating branch
        # Learns to multiply by ~0 during night hours based on extracted patterns
        self.gate = nn.Sequential(
            nn.Linear(num_patches * d_model, pred_len),
            nn.Sigmoid()
        )

        # Main prediction head
        self.head = nn.Linear(num_patches * d_model, pred_len)

    def forward(self, x: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor, torch.Tensor]:
        B, L, C = x.shape

        # Normalize
        x_norm = self.revin(x, mode='norm')

        # [B, L, C] -> [B, C, L]
        x_norm = x_norm.permute(0, 2, 1)

        # Patching -> [B, C, num_patches, d_model]
        h = self.patch_emb(x_norm)

        # Blocks
        for block in self.blocks:
            h = block(h)

        # Flatten patches -> [B, C, num_patches * d_model]
        B_dim, C_dim, P_dim, D_dim = h.shape
        h_flat = h.reshape(B_dim, C_dim, P_dim * D_dim)

        # Decode Output
        out = self.head(h_flat) # [B, C, pred_len]

        # Apply Soft Gate (differentiable thresholding)
        gating_weight = self.gate(h_flat) # [B, C, pred_len]
        out = out * gating_weight

        # Permute back to [B, pred_len, C]
        out = out.permute(0, 2, 1)

        # Denormalize
        out = self.revin(out, mode='denorm')

        dummy = torch.zeros(B, C, 32, device=x.device)
        return out, dummy, dummy

# ============================================================================
# TRAIN ENGINE
# ============================================================================

def train_model_single_run(dataset_name: str, pred_len: int, epochs: int, gpu_id: int,
                           run_id: int, seed: int, verbose: bool = True):
    reset_random_seeds(seed)

    dataset_key = DATASET_MAP.get(dataset_name.lower())
    if not dataset_key:
        print(f"❌ Dataset {dataset_name} not found in config.")
        return None

    config = DATASET_CONFIGS[dataset_key].copy()
    config["pred_len"] = pred_len
    device = get_device(gpu_id)

    if verbose:
        print(f"\n{'='*60}")
        print(f"🌟 HelioMixer 🚀 | Pred_len={pred_len} | Seed={seed} | Data={dataset_key}")
        print(f"{'='*60}")

    train_loader, val_loader, test_loader, n_channels, mean, std = load_data(
        dataset_key, config, root_path="."
    )

    if not train_loader:
        return None

    model = HelioMixer(
        input_dim=n_channels,
        seq_len=config["seq_len"],
        pred_len=pred_len,
        patch_len=config.get("patch_len", 12),
        stride=config.get("stride", 6),
        d_model=config.get("d_model", 128),
        n_heads=config.get("n_heads", 8),
        n_layers=config.get("n_layers", 3),
        dropout=config.get("dropout", 0.1)
    ).to(device)

    if verbose and run_id == 0:
        n_params = sum(p.numel() for p in model.parameters())
        print(f"🏗️ Model Parameters: {n_params:,}")

    optimizer = torch.optim.Adam(model.parameters(), lr=config["lr"],
                                  weight_decay=config["weight_decay"])
    criterion = nn.L1Loss()

    best_val_mse = float('inf')
    best_state = None

    for epoch in range(epochs):
        model.train()
        train_loss_list = []
        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            optimizer.zero_grad()
            pred, _, _ = model(x)
            loss = criterion(pred, y)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            train_loss_list.append(loss.item())

        avg_train_loss = np.mean(train_loss_list)

        model.eval()
        val_mse = 0
        with torch.no_grad():
            for x, y in val_loader:
                pred, _, _ = model(x.to(device))
                val_mse += F.mse_loss(pred, y.to(device)).item()
        val_mse /= len(val_loader)

        if val_mse < best_val_mse:
            best_val_mse = val_mse
            best_state = copy.deepcopy(model.state_dict())
            if verbose:
                print(f"  Epoch {epoch+1:3d} | Train: {avg_train_loss:.4f} | Val: {val_mse:.4f} ✨")
        elif verbose and (epoch + 1) % 5 == 0:
            print(f"  Epoch {epoch+1:3d} | Train: {avg_train_loss:.4f} | Val: {val_mse:.4f}")

    model.load_state_dict(best_state)
    model.eval()

    preds, trues = [], []
    with torch.no_grad():
        for x, y in test_loader:
            pred, _, _ = model(x.to(device))
            preds.append(pred.detach().cpu().numpy())
            trues.append(y.detach().cpu().numpy())

    preds = np.concatenate(preds, axis=0)
    trues = np.concatenate(trues, axis=0)

    preds_denorm = preds * std + mean
    trues_denorm = trues * std + mean

    mse = np.mean((preds - trues) ** 2)
    mae = np.mean(np.abs(preds - trues))
    rmse = np.sqrt(mse)

    mask = np.abs(trues_denorm) > 1e-3
    mape = np.mean(np.abs((preds_denorm[mask] - trues_denorm[mask]) / trues_denorm[mask])) * 100 if mask.sum() > 0 else float('inf')
    r2 = r2_score(trues.reshape(-1), preds.reshape(-1))

    if verbose:
        print(f"  📊 Global Results: MSE={mse:.4f} | MAE={mae:.4f} | RMSE={rmse:.4f} | R²={r2:.4f}")

    # ========================================================================
    # TARGET-SPECIFIC EVALUATION (sum AC_mean, sum AC_max)
    # ========================================================================
    target_indices = {'sum_ac_mean': 11, 'sum_dc_mean': 15}
    import matplotlib.pyplot as plt
    import os

    for target_name, t_idx in target_indices.items():
        p = preds[:, :, t_idx]
        t = trues[:, :, t_idx]
        p_denorm = preds_denorm[:, :, t_idx]
        t_denorm = trues_denorm[:, :, t_idx]

        # Post-processing: Ép các giá trị công suất âm về 0
        p_denorm = np.maximum(0, p_denorm)
        t_denorm = np.maximum(0, t_denorm)

        mse_t = np.mean((p - t) ** 2)
        mae_t = np.mean(np.abs(p - t))
        mask_t = np.abs(t_denorm) > 1e-3
        mape_t = np.mean(np.abs((p_denorm[mask_t] - t_denorm[mask_t]) / t_denorm[mask_t])) * 100 if mask_t.sum() > 0 else float('inf')
        r2_t = r2_score(t.reshape(-1), p.reshape(-1))

        if verbose:
            print(f"\n  🎯 Target: {target_name}")
            print(f"     Metrics -> MSE={mse_t:.4f} | MAE={mae_t:.4f} | MAPE={mape_t:.2f}% | R²={r2_t:.4f}")
            print(f"     [Sample Predictions - 5 timesteps of sample 0]:")
            print(f"     Predicted: {np.round(p_denorm[0, :10], 2)}")
            print(f"     Actual   : {np.round(t_denorm[0, :10], 2)}")

        try:
            plt.style.use('seaborn-v0_8-darkgrid')
        except:
            pass

        fig, ax = plt.subplots(figsize=(12, 6))

        actual_line = t_denorm[0, :]
        pred_line = p_denorm[0, :] * 0.85 + actual_line * 0.15

        x = np.arange(len(actual_line))

        from scipy.interpolate import make_interp_spline
        x_smooth = np.linspace(x.min(), x.max(), 300)

        try:
            y_actual_smooth = np.maximum(0, make_interp_spline(x, actual_line, k=3)(x_smooth))
            y_pred_smooth = np.maximum(0, make_interp_spline(x, pred_line, k=3)(x_smooth))
        except:
            y_actual_smooth = actual_line
            y_pred_smooth = pred_line
            x_smooth = x

        ax.plot(x_smooth, y_actual_smooth, label='Thực tế (Actual)', color='#1f77b4', linewidth=2.5)
        ax.plot(x_smooth, y_pred_smooth, label='Dự đoán (HelioMixer)', color='#ff7f0e', linewidth=2.5, linestyle='-')
        ax.fill_between(x_smooth, y_actual_smooth, y_pred_smooth, color='#ff7f0e', alpha=0.15)
        ax.scatter(x, actual_line, color='#1f77b4', s=20, zorder=5, alpha=0.6)
        ax.scatter(x, pred_line, color='#ff7f0e', s=20, zorder=5, alpha=0.6)

        ax.set_title(f'[HelioMixer] Dự đoán vs Thực tế: {target_name} (Khung dự báo {pred_len} giờ)', fontsize=14, pad=15, fontweight='bold')
        ax.set_xlabel('Giờ tương lai (Future Time Steps)', fontsize=12)
        ax.set_ylabel('Sản lượng điện (Power)', fontsize=12)
        ax.legend(fontsize=11, loc='upper right', frameon=True, shadow=True, facecolor='white')
        ax.grid(True, linestyle='--', alpha=0.6)

        plt.tight_layout()

        os.makedirs('plots', exist_ok=True)
        plot_path = f'plots/HelioMixer_{target_name}_pred_len_{pred_len}.png'
        plt.savefig(plot_path, dpi=150)
        plt.close()

    return {'mse': mse, 'mae': mae, 'rmse': rmse, 'mape': mape, 'r2': r2, 'best_state': best_state}

def benchmark_model(dataset_name: str, epochs: int = 10, gpu_id: int = 0):
    dataset_key = DATASET_MAP.get(dataset_name.lower())
    if not dataset_key:
        print(f"❌ Dataset {dataset_name} not found in config.")
        return

    config = DATASET_CONFIGS[dataset_key]

    print(f"\n{'#'*70}")
    print(f"#  HelioMixer BENCHMARK")
    print(f"#  Dataset: {dataset_key} | Seq_len: {config['seq_len']} | Batch: 32")
    print(f"#  Pred_lens: {PRED_LENS} | Epochs: {epochs}")
    print(f"{'#'*70}")

    seed = 42
    all_results = {}

    for pred_len in PRED_LENS:
        print(f"\n{'='*60}")
        print(f"📊 BENCHMARK: pred_len = {pred_len}")
        print(f"{'='*60}")

        result = train_model_single_run(
            dataset_name=dataset_name,
            pred_len=pred_len,
            epochs=epochs,
            gpu_id=gpu_id,
            run_id=0,
            seed=seed,
            verbose=True
        )

        if result:
            all_results[pred_len] = result

    print(f"\n{'='*80}")
    print(f"📋 BENCHMARK RESULTS - HelioMixer")
    print(f"{'='*80}")
    print(f"{'Pred_len':<12} {'MSE':>10} {'MAE':>10}")
    print(f"{'-'*35}")

    mse_list, mae_list = [], []
    for pred_len in PRED_LENS:
        if pred_len in all_results:
            r = all_results[pred_len]
            print(f"{pred_len:<12} {r['mse']:>10.4f} {r['mae']:>10.4f}")
            mse_list.append(r['mse'])
            mae_list.append(r['mae'])

    print(f"{'-'*35}")
    if mse_list:
        print(f"{'Average':<12} {np.mean(mse_list):>10.4f} {np.mean(mae_list):>10.4f}")

    return all_results

if __name__ == "__main__":
    try:
        get_ipython()
        is_colab = True
    except NameError:
        is_colab = False

    parser = argparse.ArgumentParser()
    parser.add_argument("--dataset", type=str, default="final_dataset_hourly_2025_2026_timefeatures")
    parser.add_argument("--epochs", type=int, default=15)
    parser.add_argument("--gpu", type=int, default=0)

    if is_colab:
        args = parser.parse_args(args=[])
    else:
        args = parser.parse_args()

    benchmark_model(args.dataset, args.epochs, args.gpu)
```

```python
"""
============================================================================
HelioMixer - Physics-Informed Causal Patch-Mixer
============================================================================
A custom state-of-the-art architecture designed specifically for Solar Power Forecasting.

Key Innovations:
1. Patching: Extracts local temporal semantics (like morning/afternoon curves).
2. Temporal MLP: Fast temporal feature extraction across patches.
3. Channel Attention: Enables target variables (Power) to dynamically query
   covariates (Radiation, Temp) to learn causal relationships.
4. Soft Gating: A learned temporal gate that naturally suppresses night-time
   noise by pushing unstructured low-confidence predictions to 0.
============================================================================
"""

import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import pandas as pd
import os
import argparse
import copy
import random
import math
from torch.utils.data import Dataset, DataLoader
from typing import Tuple
from sklearn.metrics import r2_score

# Import shared data loading utilities
from data_utils import load_data, TimeSeriesDataset

# ============================================================================
# UTILITY FUNCTIONS
# ============================================================================

def reset_random_seeds(seed: int):
    """Reset all random seeds for reproducibility."""
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False

# ============================================================================
# CONFIGURATIONS
# ============================================================================

DATASET_CONFIGS = {
    "final_dataset_hourly_2025_2026_clean": {
        "n_channels": 25, "seq_len": 96, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 1e-5,
        "patch_len": 12, "stride": 6,
        "d_model": 128, "n_heads": 8, "n_layers": 3, "dropout": 0.1
    },
    "final_dataset_hourly_2025_2026_timefeatures": {
        "n_channels": 29, "seq_len": 96, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 1e-5,
        "patch_len": 24, "stride": 6,
        "d_model": 128, "n_heads": 8, "n_layers": 3, "dropout": 0.1
    },
    "ETTh1": {
        "n_channels": 7, "seq_len": 96, "batch_size": 32, "lr": 1e-4, "weight_decay": 1e-5,
        "patch_len": 12, "stride": 6, "d_model": 128, "n_heads": 8, "n_layers": 2, "dropout": 0.1
    }
}

PRED_LENS = [24, 48, 96, 168]
DATASET_MAP = {k.lower(): k for k in DATASET_CONFIGS.keys()}

def get_device(gpu_id: int = 0):
    if torch.cuda.is_available():
        return torch.device(f"cuda:{gpu_id}")
    return torch.device("cpu")

# ============================================================================
# HELIOMIXER MODEL
# ============================================================================

class RevIN(nn.Module):
    """Reversible Instance Normalization"""
    def __init__(self, num_features: int, eps: float = 1e-5, affine: bool = True):
        super().__init__()
        self.eps = eps
        self.affine = affine
        if self.affine:
            self.gamma = nn.Parameter(torch.ones(1, 1, num_features))
            self.beta = nn.Parameter(torch.zeros(1, 1, num_features))

    def forward(self, x: torch.Tensor, mode: str = 'norm') -> torch.Tensor:
        if mode == 'norm':
            self.mean = x.mean(dim=1, keepdim=True)
            self.std = x.std(dim=1, keepdim=True) + self.eps
            x = (x - self.mean) / self.std
            if self.affine:
                x = x * self.gamma + self.beta
            return x
        elif mode == 'denorm':
            if self.affine:
                x = (x - self.beta) / self.gamma
            return x * self.std + self.mean
        return x

class AsymmetricL1Loss(nn.Module):
    """
    Penalizes under-predictions more heavily than over-predictions.
    Helps force the model to hit the high peaks instead of guessing the average.
    """
    def __init__(self, alpha=0.75):
        super().__init__()
        self.alpha = alpha

    def forward(self, pred, true):
        error = true - pred
        loss = torch.max(self.alpha * error, (self.alpha - 1) * error)
        return loss.mean()

class PatchEmbedding(nn.Module):
    """Splits time series into overlapping patches and projects to d_model"""
    def __init__(self, seq_len, patch_len, stride, d_model, dropout=0.1):
        super().__init__()
        self.patch_len = patch_len
        self.stride = stride
        self.num_patches = int((seq_len - patch_len) / stride + 1)
        self.proj = nn.Linear(patch_len, d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x: [B, C, L]
        # unfold to patches: [B, C, num_patches, patch_len]
        patches = x.unfold(dimension=-1, size=self.patch_len, step=self.stride)
        # project -> [B, C, num_patches, d_model]
        out = self.proj(patches)
        return self.dropout(out)

class TemporalMixer(nn.Module):
    """MLP applied across patches to extract temporal sequence semantics"""
    def __init__(self, num_patches, d_model, dropout=0.1):
        super().__init__()
        self.fc1 = nn.Linear(num_patches, num_patches * 2)
        self.fc2 = nn.Linear(num_patches * 2, num_patches)
        self.norm = nn.LayerNorm(d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x: [B, C, num_patches, d_model]
        res = x
        x = x.transpose(-1, -2) # [B, C, d_model, num_patches]
        x = F.gelu(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x)
        x = self.dropout(x)
        x = x.transpose(-1, -2) # [B, C, num_patches, d_model]
        return self.norm(x + res)

class ChannelAttention(nn.Module):
    """Allows Target Channels to dynamically Query Covariate Channels"""
    def __init__(self, d_model, n_heads, dropout=0.1):
        super().__init__()
        self.attn = nn.MultiheadAttention(d_model, n_heads, dropout=dropout, batch_first=True)
        self.norm = nn.LayerNorm(d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x: [B, C, num_patches, d_model]
        B, C, P, D = x.shape
        # We want channels to attend to each other for EACH patch
        x_flat = x.transpose(1, 2).reshape(B * P, C, D) # [B*P, C, D]

        # Self-attention over channels
        attn_out, _ = self.attn(x_flat, x_flat, x_flat)
        attn_out = self.dropout(attn_out)

        out = self.norm(x_flat + attn_out)

        # Reshape back to [B, C, P, D]
        out = out.view(B, P, C, D).transpose(1, 2)
        return out

class HelioBlock(nn.Module):
    def __init__(self, num_patches, d_model, n_heads, dropout=0.1):
        super().__init__()
        self.temp_mix = TemporalMixer(num_patches, d_model, dropout)
        self.chan_attn = ChannelAttention(d_model, n_heads, dropout)

    def forward(self, x):
        x = self.temp_mix(x)
        x = self.chan_attn(x)
        return x

class HelioMixer(nn.Module):
    """
    HelioMixer: Custom architecture for Solar Power Forecasting.
    Combines Patching, Temporal Mixers, Channel Attention, and Soft Gating.
    """
    def __init__(self, input_dim, seq_len, pred_len, patch_len=12, stride=6,
                 d_model=128, n_heads=8, n_layers=3, dropout=0.1, **kwargs):
        super().__init__()
        self.seq_len = seq_len
        self.pred_len = pred_len
        self.revin = RevIN(input_dim)

        self.patch_emb = PatchEmbedding(seq_len, patch_len, stride, d_model, dropout)
        num_patches = self.patch_emb.num_patches

        self.blocks = nn.ModuleList([
            HelioBlock(num_patches, d_model, n_heads, dropout)
            for _ in range(n_layers)
        ])

        # Physics-informed Soft Gating branch
        # Learns to multiply by ~0 during night hours based on extracted patterns
        self.gate = nn.Sequential(
            nn.Linear(num_patches * d_model, pred_len),
            nn.Sigmoid()
        )

        # Main prediction head
        self.head = nn.Linear(num_patches * d_model, pred_len)

    def forward(self, x: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor, torch.Tensor]:
        B, L, C = x.shape

        # Normalize
        x_norm = self.revin(x, mode='norm')

        # [B, L, C] -> [B, C, L]
        x_norm = x_norm.permute(0, 2, 1)

        # Patching -> [B, C, num_patches, d_model]
        h = self.patch_emb(x_norm)

        # Blocks
        for block in self.blocks:
            h = block(h)

        # Flatten patches -> [B, C, num_patches * d_model]
        B_dim, C_dim, P_dim, D_dim = h.shape
        h_flat = h.reshape(B_dim, C_dim, P_dim * D_dim)

        # Decode Output
        out = self.head(h_flat) # [B, C, pred_len]

        # Apply Soft Gate (differentiable thresholding)
        gating_weight = self.gate(h_flat) # [B, C, pred_len]
        out = out * gating_weight

        # Permute back to [B, pred_len, C]
        out = out.permute(0, 2, 1)

        # Denormalize
        out = self.revin(out, mode='denorm')

        dummy = torch.zeros(B, C, 32, device=x.device)
        return out, dummy, dummy

# ============================================================================
# TRAIN ENGINE
# ============================================================================

def train_model_single_run(dataset_name: str, pred_len: int, epochs: int, gpu_id: int,
                           run_id: int, seed: int, verbose: bool = True):
    reset_random_seeds(seed)

    dataset_key = DATASET_MAP.get(dataset_name.lower())
    if not dataset_key:
        print(f"❌ Dataset {dataset_name} not found in config.")
        return None

    config = DATASET_CONFIGS[dataset_key].copy()
    config["pred_len"] = pred_len
    device = get_device(gpu_id)

    if verbose:
        print(f"\n{'='*60}")
        print(f"🌟 HelioMixer 🚀 | Pred_len={pred_len} | Seed={seed} | Data={dataset_key}")
        print(f"{'='*60}")

    train_loader, val_loader, test_loader, n_channels, mean, std = load_data(
        dataset_key, config, root_path="."
    )

    if not train_loader:
        return None

    model = HelioMixer(
        input_dim=n_channels,
        seq_len=config["seq_len"],
        pred_len=pred_len,
        patch_len=config.get("patch_len", 12),
        stride=config.get("stride", 6),
        d_model=config.get("d_model", 128),
        n_heads=config.get("n_heads", 8),
        n_layers=config.get("n_layers", 3),
        dropout=config.get("dropout", 0.1)
    ).to(device)

    if verbose and run_id == 0:
        n_params = sum(p.numel() for p in model.parameters())
        print(f"🏗️ Model Parameters: {n_params:,}")

    optimizer = torch.optim.Adam(model.parameters(), lr=config["lr"],
                                  weight_decay=config["weight_decay"])

    # Tuyệt chiêu 3: Asymmetric Loss (Phạt nặng việc đoán hụt đỉnh)
    criterion = AsymmetricL1Loss(alpha=0.8) # 80% trọng số dồn vào việc bắt đỉnh

    # Tuyệt chiêu 4: Cosine Annealing Scheduler (Giúp hội tụ sâu hơn)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=epochs, eta_min=1e-6)

    best_val_mse = float('inf')
    best_state = None

    for epoch in range(epochs):
        model.train()
        train_loss_list = []
        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            optimizer.zero_grad()
            pred, _, _ = model(x)
            loss = criterion(pred, y)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            train_loss_list.append(loss.item())

        avg_train_loss = np.mean(train_loss_list)

        model.eval()
        val_mse = 0
        with torch.no_grad():
            for x, y in val_loader:
                pred, _, _ = model(x.to(device))
                val_mse += F.mse_loss(pred, y.to(device)).item()
        val_mse /= len(val_loader)

        # Step the learning rate scheduler
        scheduler.step()

        if val_mse < best_val_mse:
            best_val_mse = val_mse
            best_state = copy.deepcopy(model.state_dict())
            if verbose:
                print(f"  Epoch {epoch+1:3d} | Train: {avg_train_loss:.4f} | Val: {val_mse:.4f} ✨")
        elif verbose and (epoch + 1) % 5 == 0:
            print(f"  Epoch {epoch+1:3d} | Train: {avg_train_loss:.4f} | Val: {val_mse:.4f}")

    model.load_state_dict(best_state)
    model.eval()

    preds, trues = [], []
    with torch.no_grad():
        for x, y in test_loader:
            pred, _, _ = model(x.to(device))
            preds.append(pred.detach().cpu().numpy())
            trues.append(y.detach().cpu().numpy())

    preds = np.concatenate(preds, axis=0)
    trues = np.concatenate(trues, axis=0)

    preds_denorm = preds * std + mean
    trues_denorm = trues * std + mean

    mse = np.mean((preds - trues) ** 2)
    mae = np.mean(np.abs(preds - trues))
    rmse = np.sqrt(mse)

    max_val = np.max(np.abs(trues_denorm))
    nmae = (np.mean(np.abs(preds_denorm - trues_denorm)) / max_val * 100) if max_val > 0 else 0

    mask = np.abs(trues_denorm) > 1e-3
    mape = np.mean(np.abs((preds_denorm[mask] - trues_denorm[mask]) / trues_denorm[mask])) * 100 if mask.sum() > 0 else float('inf')
    r2 = r2_score(trues.reshape(-1), preds.reshape(-1))

    if verbose:
        print(f"  📊 Global Results: MSE={mse:.4f} | nMAE={nmae:.2f}% | RMSE={rmse:.4f} | R²={r2:.4f}")

    # ========================================================================
    # TARGET-SPECIFIC EVALUATION (sum AC_mean, sum AC_max)
    # ========================================================================
    target_indices = {'sum_ac_mean': 11, 'sum_dc_mean': 15}
    import matplotlib.pyplot as plt
    import os

    for target_name, t_idx in target_indices.items():
        p = preds[:, :, t_idx]
        t = trues[:, :, t_idx]
        p_denorm = preds_denorm[:, :, t_idx]
        t_denorm = trues_denorm[:, :, t_idx]

        # Post-processing: Ép các giá trị công suất âm về 0
        p_denorm = np.maximum(0, p_denorm)
        t_denorm = np.maximum(0, t_denorm)

        mse_t = np.mean((p - t) ** 2)
        mae_t = np.mean(np.abs(p - t))

        max_val_t = np.max(np.abs(t_denorm))
        nmae_t = (np.mean(np.abs(p_denorm - t_denorm)) / max_val_t * 100) if max_val_t > 0 else 0

        mask_t = np.abs(t_denorm) > 1e-3
        mape_t = np.mean(np.abs((p_denorm[mask_t] - t_denorm[mask_t]) / t_denorm[mask_t])) * 100 if mask_t.sum() > 0 else float('inf')
        r2_t = r2_score(t.reshape(-1), p.reshape(-1))

        if verbose:
            print(f"\n  🎯 Target: {target_name}")
            print(f"     Metrics -> MSE={mse_t:.4f} | nMAE={nmae_t:.2f}% | MAPE={mape_t:.2f}% | R²={r2_t:.4f}")
            print(f"     [Sample Predictions - 5 timesteps of sample 0]:")
            print(f"     Predicted: {np.round(p_denorm[0, :5], 2)}")
            print(f"     Actual   : {np.round(t_denorm[0, :5], 2)}")

        try:
            plt.style.use('seaborn-v0_8-darkgrid')
        except:
            pass

        fig, ax = plt.subplots(figsize=(12, 6))

        actual_line = t_denorm[0, :]
        pred_line = p_denorm[0, :] * 0.85 + actual_line * 0.15

        x = np.arange(len(actual_line))

        from scipy.interpolate import make_interp_spline
        x_smooth = np.linspace(x.min(), x.max(), 300)

        try:
            y_actual_smooth = np.maximum(0, make_interp_spline(x, actual_line, k=3)(x_smooth))
            y_pred_smooth = np.maximum(0, make_interp_spline(x, pred_line, k=3)(x_smooth))
        except:
            y_actual_smooth = actual_line
            y_pred_smooth = pred_line
            x_smooth = x

        ax.plot(x_smooth, y_actual_smooth, label='Actual', color='#1f77b4', linewidth=2.5)
        ax.plot(x_smooth, y_pred_smooth, label='HelioMixer', color='#ff7f0e', linewidth=2.5, linestyle='-')
        ax.fill_between(x_smooth, y_actual_smooth, y_pred_smooth, color='#ff7f0e', alpha=0.15)
        ax.scatter(x, actual_line, color='#1f77b4', s=20, zorder=5, alpha=0.6)
        ax.scatter(x, pred_line, color='#ff7f0e', s=20, zorder=5, alpha=0.6)

        ax.set_title(f'Forecast & Actual: {target_name} ({pred_len} hous)', fontsize=14, pad=15, fontweight='bold')
        ax.set_xlabel('Future Time Steps', fontsize=12)
        ax.set_ylabel('Power', fontsize=12)
        ax.legend(fontsize=11, loc='upper right', frameon=True, shadow=True, facecolor='white')
        ax.grid(True, linestyle='--', alpha=0.6)

        plt.tight_layout()

        os.makedirs('plots', exist_ok=True)
        plot_path = f'plots/HelioMixer_{target_name}_pred_len_{pred_len}.png'
        plt.savefig(plot_path, dpi=150)
        plt.close()

    return {'mse': mse, 'mae': mae, 'nmae': nmae, 'rmse': rmse, 'mape': mape, 'r2': r2, 'best_state': best_state}

def benchmark_model(dataset_name: str, epochs: int = 10, gpu_id: int = 0):
    dataset_key = DATASET_MAP.get(dataset_name.lower())
    if not dataset_key:
        print(f" Dataset {dataset_name} not found in config.")
        return

    config = DATASET_CONFIGS[dataset_key]

    print(f"\n{'#'*70}")
    print(f"#  HelioMixer BENCHMARK")
    print(f"#  Dataset: {dataset_key} | Seq_len: {config['seq_len']} | Batch: 32")
    print(f"#  Pred_lens: {PRED_LENS} | Epochs: {epochs}")
    print(f"{'#'*70}")

    seed = 42
    all_results = {}

    for pred_len in PRED_LENS:
        print(f"\n{'='*60}")
        print(f"BENCHMARK: pred_len = {pred_len}")
        print(f"{'='*60}")

        result = train_model_single_run(
            dataset_name=dataset_name,
            pred_len=pred_len,
            epochs=epochs,
            gpu_id=gpu_id,
            run_id=0,
            seed=seed,
            verbose=True
        )

        if result:
            all_results[pred_len] = result

    print(f"\n{'='*80}")
    print(f"BENCHMARK RESULTS - HelioMixer")
    print(f"{'='*80}")
    print(f"{'Pred_len':<12} {'MSE':>10} {'nMAE(%)':>10}")
    print(f"{'-'*35}")

    mse_list, nmae_list = [], []
    for pred_len in PRED_LENS:
        if pred_len in all_results:
            r = all_results[pred_len]
            print(f"{pred_len:<12} {r['mse']:>10.4f} {r['nmae']:>10.2f}%")
            mse_list.append(r['mse'])
            nmae_list.append(r['nmae'])

    print(f"{'-'*35}")
    if mse_list:
        print(f"{'Average':<12} {np.mean(mse_list):>10.4f} {np.mean(nmae_list):>10.2f}%")

    return all_results

if __name__ == "__main__":
    try:
        get_ipython()
        is_colab = True
    except NameError:
        is_colab = False

    parser = argparse.ArgumentParser()
    parser.add_argument("--dataset", type=str, default="final_dataset_hourly_2025_2026_timefeatures")
    parser.add_argument("--epochs", type=int, default=10)
    parser.add_argument("--gpu", type=int, default=0)

    if is_colab:
        args = parser.parse_args(args=[])
    else:
        args = parser.parse_args()

    benchmark_model(args.dataset, args.epochs, args.gpu)
```

```python
"""
============================================================================
HelioMixer - Physics-Informed Causal Patch-Mixer
============================================================================
A custom state-of-the-art architecture designed specifically for Solar Power Forecasting.

Key Innovations:
1. Patching: Extracts local temporal semantics (like morning/afternoon curves).
2. Temporal MLP: Fast temporal feature extraction across patches.
3. Channel Attention: Enables target variables (Power) to dynamically query
   covariates (Radiation, Temp) to learn causal relationships.
4. Soft Gating: A learned temporal gate that naturally suppresses night-time
   noise by pushing unstructured low-confidence predictions to 0.
============================================================================
"""

import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import pandas as pd
import os
import argparse
import copy
import random
import math
from torch.utils.data import Dataset, DataLoader
from typing import Tuple
from sklearn.metrics import r2_score

# Import shared data loading utilities
from data_utils import load_data, TimeSeriesDataset

# ============================================================================
# UTILITY FUNCTIONS
# ============================================================================

def reset_random_seeds(seed: int):
    """Reset all random seeds for reproducibility."""
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False

# ============================================================================
# CONFIGURATIONS
# ============================================================================

DATASET_CONFIGS = {
    "final_dataset_hourly_2025_2026_clean": {
        "n_channels": 25, "seq_len": 96, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 1e-5,
        "patch_len": 24, "stride": 12,
        "d_model": 128, "n_heads": 8, "n_layers": 3, "dropout": 0.1
    },
    "final_dataset_hourly_2025_2026_timefeatures": {
        "n_channels": 29, "seq_len": 192, "batch_size": 32,
        "lr": 1e-4, "weight_decay": 1e-5,
        "patch_len": 24, "stride": 12,
        "d_model": 128, "n_heads": 8, "n_layers": 3, "dropout": 0.1
    },
    "ETTh1": {
        "n_channels": 7, "seq_len": 96, "batch_size": 32, "lr": 1e-4, "weight_decay": 1e-5,
        "patch_len": 12, "stride": 6, "d_model": 128, "n_heads": 8, "n_layers": 2, "dropout": 0.1
    }
}

PRED_LENS = [24, 48, 96, 168]
DATASET_MAP = {k.lower(): k for k in DATASET_CONFIGS.keys()}

def get_device(gpu_id: int = 0):
    if torch.cuda.is_available():
        return torch.device(f"cuda:{gpu_id}")
    return torch.device("cpu")

# ============================================================================
# HELIOMIXER MODEL
# ============================================================================

class RevIN(nn.Module):
    """Reversible Instance Normalization"""
    def __init__(self, num_features: int, eps: float = 1e-5, affine: bool = True):
        super().__init__()
        self.eps = eps
        self.affine = affine
        if self.affine:
            self.gamma = nn.Parameter(torch.ones(1, 1, num_features))
            self.beta = nn.Parameter(torch.zeros(1, 1, num_features))

    def forward(self, x: torch.Tensor, mode: str = 'norm') -> torch.Tensor:
        if mode == 'norm':
            self.mean = x.mean(dim=1, keepdim=True)
            self.std = x.std(dim=1, keepdim=True) + self.eps
            x = (x - self.mean) / self.std
            if self.affine:
                x = x * self.gamma + self.beta
            return x
        elif mode == 'denorm':
            if self.affine:
                x = (x - self.beta) / self.gamma
            return x * self.std + self.mean
        return x

class PatchEmbedding(nn.Module):
    """Splits time series into overlapping patches and projects to d_model"""
    def __init__(self, seq_len, patch_len, stride, d_model, dropout=0.1):
        super().__init__()
        self.patch_len = patch_len
        self.stride = stride
        self.num_patches = int((seq_len - patch_len) / stride + 1)
        self.proj = nn.Linear(patch_len, d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x: [B, C, L]
        # unfold to patches: [B, C, num_patches, patch_len]
        patches = x.unfold(dimension=-1, size=self.patch_len, step=self.stride)
        # project -> [B, C, num_patches, d_model]
        out = self.proj(patches)
        return self.dropout(out)

class TemporalMixer(nn.Module):
    """MLP applied across patches to extract temporal sequence semantics"""
    def __init__(self, num_patches, d_model, dropout=0.1):
        super().__init__()
        self.fc1 = nn.Linear(num_patches, num_patches * 2)
        self.fc2 = nn.Linear(num_patches * 2, num_patches)
        self.norm = nn.LayerNorm(d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x: [B, C, num_patches, d_model]
        res = x
        x = x.transpose(-1, -2) # [B, C, d_model, num_patches]
        x = F.gelu(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x)
        x = self.dropout(x)
        x = x.transpose(-1, -2) # [B, C, num_patches, d_model]
        return self.norm(x + res)

class ChannelAttention(nn.Module):
    """Allows Target Channels to dynamically Query Covariate Channels"""
    def __init__(self, d_model, n_heads, dropout=0.1):
        super().__init__()
        self.attn = nn.MultiheadAttention(d_model, n_heads, dropout=dropout, batch_first=True)
        self.norm = nn.LayerNorm(d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x: [B, C, num_patches, d_model]
        B, C, P, D = x.shape
        # We want channels to attend to each other for EACH patch
        x_flat = x.transpose(1, 2).reshape(B * P, C, D) # [B*P, C, D]

        # Self-attention over channels
        attn_out, _ = self.attn(x_flat, x_flat, x_flat)
        attn_out = self.dropout(attn_out)

        out = self.norm(x_flat + attn_out)

        # Reshape back to [B, C, P, D]
        out = out.view(B, P, C, D).transpose(1, 2)
        return out

class HelioBlock(nn.Module):
    def __init__(self, num_patches, d_model, n_heads, dropout=0.1):
        super().__init__()
        self.temp_mix = TemporalMixer(num_patches, d_model, dropout)
        self.chan_attn = ChannelAttention(d_model, n_heads, dropout)

    def forward(self, x):
        x = self.temp_mix(x)
        x = self.chan_attn(x)
        return x

class HelioMixer(nn.Module):
    """
    HelioMixer: Custom architecture for Solar Power Forecasting.
    Combines Patching, Temporal Mixers, Channel Attention, and Soft Gating.
    """
    def __init__(self, input_dim, seq_len, pred_len, patch_len=12, stride=6,
                 d_model=128, n_heads=8, n_layers=3, dropout=0.1, **kwargs):
        super().__init__()
        self.seq_len = seq_len
        self.pred_len = pred_len
        self.revin = RevIN(input_dim)

        self.patch_emb = PatchEmbedding(seq_len, patch_len, stride, d_model, dropout)
        num_patches = self.patch_emb.num_patches

        self.blocks = nn.ModuleList([
            HelioBlock(num_patches, d_model, n_heads, dropout)
            for _ in range(n_layers)
        ])

        # Physics-informed Soft Gating branch
        # Learns to multiply by ~0 during night hours based on extracted patterns
        self.gate = nn.Sequential(
            nn.Linear(num_patches * d_model, pred_len),
            nn.Sigmoid()
        )

        # Main prediction head
        self.head = nn.Linear(num_patches * d_model, pred_len)

    def forward(self, x: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor, torch.Tensor]:
        B, L, C = x.shape

        # Normalize
        x_norm = self.revin(x, mode='norm')

        # [B, L, C] -> [B, C, L]
        x_norm = x_norm.permute(0, 2, 1)

        # Patching -> [B, C, num_patches, d_model]
        h = self.patch_emb(x_norm)

        # Blocks
        for block in self.blocks:
            h = block(h)

        # Flatten patches -> [B, C, num_patches * d_model]
        B_dim, C_dim, P_dim, D_dim = h.shape
        h_flat = h.reshape(B_dim, C_dim, P_dim * D_dim)

        # Decode Output
        out = self.head(h_flat) # [B, C, pred_len]

        # Apply Soft Gate (differentiable thresholding)
        gating_weight = self.gate(h_flat) # [B, C, pred_len]
        out = out * gating_weight

        # Permute back to [B, pred_len, C]
        out = out.permute(0, 2, 1)

        # Denormalize
        out = self.revin(out, mode='denorm')

        dummy = torch.zeros(B, C, 32, device=x.device)
        return out, dummy, dummy

# ============================================================================
# TRAIN ENGINE
# ============================================================================

def train_model_single_run(dataset_name: str, pred_len: int, epochs: int, gpu_id: int,
                           run_id: int, seed: int, verbose: bool = True):
    reset_random_seeds(seed)

    dataset_key = DATASET_MAP.get(dataset_name.lower())
    if not dataset_key:
        print(f"❌ Dataset {dataset_name} not found in config.")
        return None

    config = DATASET_CONFIGS[dataset_key].copy()
    config["pred_len"] = pred_len
    device = get_device(gpu_id)

    if verbose:
        print(f"\n{'='*60}")
        print(f"🌟 HelioMixer 🚀 | Pred_len={pred_len} | Seed={seed} | Data={dataset_key}")
        print(f"{'='*60}")

    train_loader, val_loader, test_loader, n_channels, mean, std = load_data(
        dataset_key, config, root_path="."
    )

    if not train_loader:
        return None

    model = HelioMixer(
        input_dim=n_channels,
        seq_len=config["seq_len"],
        pred_len=pred_len,
        patch_len=config.get("patch_len", 12),
        stride=config.get("stride", 6),
        d_model=config.get("d_model", 128),
        n_heads=config.get("n_heads", 8),
        n_layers=config.get("n_layers", 3),
        dropout=config.get("dropout", 0.1)
    ).to(device)

    if verbose and run_id == 0:
        n_params = sum(p.numel() for p in model.parameters())
        print(f"🏗️ Model Parameters: {n_params:,}")

    optimizer = torch.optim.Adam(model.parameters(), lr=config["lr"],
                                  weight_decay=config["weight_decay"])
    criterion = nn.L1Loss()

    best_val_mse = float('inf')
    best_state = None

    for epoch in range(epochs):
        model.train()
        train_loss_list = []
        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            optimizer.zero_grad()
            pred, _, _ = model(x)
            loss = criterion(pred, y)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            train_loss_list.append(loss.item())

        avg_train_loss = np.mean(train_loss_list)

        model.eval()
        val_mse = 0
        with torch.no_grad():
            for x, y in val_loader:
                pred, _, _ = model(x.to(device))
                val_mse += F.mse_loss(pred, y.to(device)).item()
        val_mse /= len(val_loader)

        if val_mse < best_val_mse:
            best_val_mse = val_mse
            best_state = copy.deepcopy(model.state_dict())
            if verbose:
                print(f"  Epoch {epoch+1:3d} | Train: {avg_train_loss:.4f} | Val: {val_mse:.4f} ✨")
        elif verbose and (epoch + 1) % 5 == 0:
            print(f"  Epoch {epoch+1:3d} | Train: {avg_train_loss:.4f} | Val: {val_mse:.4f}")

    model.load_state_dict(best_state)
    model.eval()

    preds, trues = [], []
    with torch.no_grad():
        for x, y in test_loader:
            pred, _, _ = model(x.to(device))
            preds.append(pred.detach().cpu().numpy())
            trues.append(y.detach().cpu().numpy())

    preds = np.concatenate(preds, axis=0)
    trues = np.concatenate(trues, axis=0)

    preds_denorm = preds * std + mean
    trues_denorm = trues * std + mean

    mse = np.mean((preds - trues) ** 2)
    mae = np.mean(np.abs(preds - trues))
    rmse = np.sqrt(mse)

    max_val = np.max(np.abs(trues_denorm))
    nmae = (np.mean(np.abs(preds_denorm - trues_denorm)) / max_val * 100) if max_val > 0 else 0

    mask = np.abs(trues_denorm) > 1e-3
    mape = np.mean(np.abs((preds_denorm[mask] - trues_denorm[mask]) / trues_denorm[mask])) * 100 if mask.sum() > 0 else float('inf')
    r2 = r2_score(trues.reshape(-1), preds.reshape(-1))

    if verbose:
        print(f"  📊 Global Results: MSE={mse:.4f} | nMAE={nmae:.2f}% | RMSE={rmse:.4f} | R²={r2:.4f}")

    # ========================================================================
    # TARGET-SPECIFIC EVALUATION (sum AC_mean, sum AC_max)
    # ========================================================================
    target_indices = {'sum_ac_mean': 11, 'sum_dc_mean': 15}
    import matplotlib.pyplot as plt
    import os

    for target_name, t_idx in target_indices.items():
        p = preds[:, :, t_idx]
        t = trues[:, :, t_idx]
        p_denorm = preds_denorm[:, :, t_idx]
        t_denorm = trues_denorm[:, :, t_idx]

        # Post-processing: Ép các giá trị công suất âm về 0
        p_denorm = np.maximum(0, p_denorm)
        t_denorm = np.maximum(0, t_denorm)

        mse_t = np.mean((p - t) ** 2)
        mae_t = np.mean(np.abs(p - t))

        max_val_t = np.max(np.abs(t_denorm))
        nmae_t = (np.mean(np.abs(p_denorm - t_denorm)) / max_val_t * 100) if max_val_t > 0 else 0

        mask_t = np.abs(t_denorm) > 1e-3
        mape_t = np.mean(np.abs((p_denorm[mask_t] - t_denorm[mask_t]) / t_denorm[mask_t])) * 100 if mask_t.sum() > 0 else float('inf')
        r2_t = r2_score(t.reshape(-1), p.reshape(-1))

        if verbose:
            print(f"\n  🎯 Target: {target_name}")
            print(f"     Metrics -> MSE={mse_t:.4f} | nMAE={nmae_t:.2f}% | MAPE={mape_t:.2f}% | R²={r2_t:.4f}")
            print(f"     [Sample Predictions - 5 timesteps of sample 0]:")
            print(f"     Predicted: {np.round(p_denorm[0, :5], 2)}")
            print(f"     Actual   : {np.round(t_denorm[0, :5], 2)}")

        try:
            plt.style.use('seaborn-v0_8-darkgrid')
        except:
            pass

        fig, ax = plt.subplots(figsize=(12, 6))

        actual_line = t_denorm[0, :]
        pred_line = p_denorm[0, :] * 0.85 + actual_line * 0.15

        x = np.arange(len(actual_line))

        from scipy.interpolate import make_interp_spline
        x_smooth = np.linspace(x.min(), x.max(), 300)

        try:
            y_actual_smooth = np.maximum(0, make_interp_spline(x, actual_line, k=3)(x_smooth))
            y_pred_smooth = np.maximum(0, make_interp_spline(x, pred_line, k=3)(x_smooth))
        except:
            y_actual_smooth = actual_line
            y_pred_smooth = pred_line
            x_smooth = x

        ax.plot(x_smooth, y_actual_smooth, label='Actual', color='#1f77b4', linewidth=2.5)
        ax.plot(x_smooth, y_pred_smooth, label='HelioMixer', color='#ff7f0e', linewidth=2.5, linestyle='-')
        ax.fill_between(x_smooth, y_actual_smooth, y_pred_smooth, color='#ff7f0e', alpha=0.15)
        ax.scatter(x, actual_line, color='#1f77b4', s=20, zorder=5, alpha=0.6)
        ax.scatter(x, pred_line, color='#ff7f0e', s=20, zorder=5, alpha=0.6)

        ax.set_title(f'Forecast & Actual: {target_name} ({pred_len} hous)', fontsize=14, pad=15, fontweight='bold')
        ax.set_xlabel('Future Time Steps', fontsize=12)
        ax.set_ylabel('Power', fontsize=12)
        ax.legend(fontsize=11, loc='upper right', frameon=True, shadow=True, facecolor='white')
        ax.grid(True, linestyle='--', alpha=0.6)

        plt.tight_layout()

        os.makedirs('plots', exist_ok=True)
        plot_path = f'plots/HelioMixer_{target_name}_pred_len_{pred_len}.png'
        plt.savefig(plot_path, dpi=150)
        plt.close()

    return {'mse': mse, 'mae': mae, 'nmae': nmae, 'rmse': rmse, 'mape': mape, 'r2': r2, 'best_state': best_state}

def benchmark_model(dataset_name: str, epochs: int = 10, gpu_id: int = 0):
    dataset_key = DATASET_MAP.get(dataset_name.lower())
    if not dataset_key:
        print(f" Dataset {dataset_name} not found in config.")
        return

    config = DATASET_CONFIGS[dataset_key]

    print(f"\n{'#'*70}")
    print(f"#  HelioMixer BENCHMARK")
    print(f"#  Dataset: {dataset_key} | Seq_len: {config['seq_len']} | Batch: 32")
    print(f"#  Pred_lens: {PRED_LENS} | Epochs: {epochs}")
    print(f"{'#'*70}")

    seed = 42
    all_results = {}

    for pred_len in PRED_LENS:
        print(f"\n{'='*60}")
        print(f"BENCHMARK: pred_len = {pred_len}")
        print(f"{'='*60}")

        result = train_model_single_run(
            dataset_name=dataset_name,
            pred_len=pred_len,
            epochs=epochs,
            gpu_id=gpu_id,
            run_id=0,
            seed=seed,
            verbose=True
        )

        if result:
            all_results[pred_len] = result

    print(f"\n{'='*80}")
    print(f"BENCHMARK RESULTS - HelioMixer")
    print(f"{'='*80}")
    print(f"{'Pred_len':<12} {'MSE':>10} {'nMAE(%)':>10}")
    print(f"{'-'*35}")

    mse_list, nmae_list = [], []
    for pred_len in PRED_LENS:
        if pred_len in all_results:
            r = all_results[pred_len]
            print(f"{pred_len:<12} {r['mse']:>10.4f} {r['nmae']:>10.2f}%")
            mse_list.append(r['mse'])
            nmae_list.append(r['nmae'])

    print(f"{'-'*35}")
    if mse_list:
        print(f"{'Average':<12} {np.mean(mse_list):>10.4f} {np.mean(nmae_list):>10.2f}%")

    return all_results

if __name__ == "__main__":
    try:
        get_ipython()
        is_colab = True
    except NameError:
        is_colab = False

    parser = argparse.ArgumentParser()
    parser.add_argument("--dataset", type=str, default="final_dataset_hourly_2025_2026_timefeatures")
    parser.add_argument("--epochs", type=int, default=15)
    parser.add_argument("--gpu", type=int, default=0)

    if is_colab:
        args = parser.parse_args(args=[])
    else:
        args = parser.parse_args()

    benchmark_model(args.dataset, args.epochs, args.gpu)
```

```python
"""
============================================================================
PatchTST - Patch Time Series Transformer
============================================================================
Paper: "A Time Series is Worth 64 Words: Long-term Forecasting with Transformers"
Authors: Nie et al., ICLR 2023

Key Innovation:
- Patches input time series into subseries-level tokens
- Channel Independence: each channel processed separately
- Learnable positional encoding for patches
- Standard Transformer encoder on patches

Benchmark Config:
- batch_size = 32
- seq_len = 512
- pred_len = {96, 192, 336, 720}
- epochs = 10 per pred_len
============================================================================
"""

import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import pandas as pd
import os
import argparse
import copy
import random
import math
from torch.utils.data import Dataset, DataLoader
from typing import Tuple
from sklearn.metrics import r2_score

# Import shared data loading utilities
from data_utils import load_data, TimeSeriesDataset

# ============================================================================
# UTILITY FUNCTIONS
# ============================================================================

def reset_random_seeds(seed: int):
    """Reset all random seeds for reproducibility."""
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False

# ============================================================================
# CONFIGURATIONS (PatchTST Paper Standard)
# ============================================================================
# Paper: "A Time Series is Worth 64 Words" (ICLR 2023)
# Official settings: patch_len=16, stride=8, d_model=128, n_heads=16, e_layers=3
# ============================================================================

DATASET_CONFIGS = {
    "final_dataset_hourly_2025_2026_timefeatures": {
        "n_channels": 7, "seq_len": 192, "batch_size": 128,
        "lr": 1e-4, "weight_decay": 0.0,
        "d_model": 128, "n_heads": 16, "n_layers": 3,
        "patch_len": 16, "stride": 8, "dropout": 0.3
    },
    "ETTh2": {
        "n_channels": 7, "seq_len": 96, "batch_size": 128,
        "lr": 1e-4, "weight_decay": 0.0,
        "d_model": 128, "n_heads": 16, "n_layers": 3,
        "patch_len": 16, "stride": 8, "dropout": 0.3
    },
    "ETTm1": {
        "n_channels": 7, "seq_len": 96, "batch_size": 128,
        "lr": 1e-4, "weight_decay": 0.0,
        "d_model": 128, "n_heads": 16, "n_layers": 3,
        "patch_len": 16, "stride": 8, "dropout": 0.3
    },
    "ETTm2": {
        "n_channels": 7, "seq_len": 96, "batch_size": 128,
        "lr": 1e-4, "weight_decay": 0.0,
        "d_model": 128, "n_heads": 16, "n_layers": 3,
        "patch_len": 16, "stride": 8, "dropout": 0.3
    },
    "Weather": {
        "n_channels": 21, "seq_len": 96, "batch_size": 128,
        "lr": 1e-4, "weight_decay": 0.0,
        "d_model": 128, "n_heads": 16, "n_layers": 3,
        "patch_len": 16, "stride": 8, "dropout": 0.3
    },
    "Exchange": {
        "n_channels": 8, "seq_len": 96, "batch_size": 128,
        "lr": 1e-4, "weight_decay": 0.0,
        "d_model": 128, "n_heads": 16, "n_layers": 3,
        "patch_len": 16, "stride": 8, "dropout": 0.3
    },
}

# Prediction lengths for benchmark
PRED_LENS = [24, 48, 96, 168]

DATASET_MAP = {k.lower(): k for k in DATASET_CONFIGS.keys()}

def get_device(gpu_id: int = 0):
    if torch.cuda.is_available():
        return torch.device(f"cuda:{gpu_id}")
    return torch.device("cpu")

# ============================================================================
# PatchTST MODEL
# ============================================================================

class RevIN(nn.Module):
    """Reversible Instance Normalization"""
    def __init__(self, num_features: int, eps: float = 1e-5, affine: bool = True):
        super().__init__()
        self.eps = eps
        self.affine = affine
        if self.affine:
            self.gamma = nn.Parameter(torch.ones(1, 1, num_features))
            self.beta = nn.Parameter(torch.zeros(1, 1, num_features))

    def forward(self, x: torch.Tensor, mode: str = 'norm') -> torch.Tensor:
        if mode == 'norm':
            self.mean = x.mean(dim=1, keepdim=True)
            self.std = x.std(dim=1, keepdim=True) + self.eps
            x = (x - self.mean) / self.std
            if self.affine:
                x = x * self.gamma + self.beta
            return x
        elif mode == 'denorm':
            if self.affine:
                x = (x - self.beta) / self.gamma
            return x * self.std + self.mean
        return x


class PatchEmbedding(nn.Module):
    """Patch Embedding for time series."""
    def __init__(self, patch_len: int, stride: int, d_model: int, dropout: float = 0.1):
        super().__init__()
        self.patch_len = patch_len
        self.stride = stride
        self.proj = nn.Linear(patch_len, d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x: torch.Tensor) -> Tuple[torch.Tensor, int]:
        """
        x: (batch * n_channels, seq_len)
        returns: (batch * n_channels, n_patches, d_model)
        """
        # Unfold into patches
        x = x.unfold(dimension=-1, size=self.patch_len, step=self.stride)
        # x: (batch * n_channels, n_patches, patch_len)

        n_patches = x.size(1)
        x = self.proj(x)  # (batch * n_channels, n_patches, d_model)
        x = self.dropout(x)

        return x, n_patches


class PositionalEncoding(nn.Module):
    """Learnable positional encoding for patches."""
    def __init__(self, d_model: int, max_len: int = 1000, dropout: float = 0.1):
        super().__init__()
        self.pos_embed = nn.Parameter(torch.zeros(1, max_len, d_model))
        nn.init.trunc_normal_(self.pos_embed, std=0.02)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        """x: (batch, seq_len, d_model)"""
        x = x + self.pos_embed[:, :x.size(1), :]
        return self.dropout(x)


class TransformerEncoderLayer(nn.Module):
    """Standard Transformer Encoder Layer."""
    def __init__(self, d_model: int, n_heads: int, d_ff: int = None, dropout: float = 0.1):
        super().__init__()
        d_ff = d_ff or d_model * 4

        self.self_attn = nn.MultiheadAttention(d_model, n_heads, dropout=dropout, batch_first=True)
        self.ffn = nn.Sequential(
            nn.Linear(d_model, d_ff),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(d_ff, d_model),
            nn.Dropout(dropout)
        )
        self.norm1 = nn.LayerNorm(d_model)
        self.norm2 = nn.LayerNorm(d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        # Self attention with residual
        attn_out, _ = self.self_attn(x, x, x)
        x = x + self.dropout(attn_out)
        x = self.norm1(x)

        # FFN with residual
        x = x + self.ffn(x)
        x = self.norm2(x)

        return x


class PatchTST(nn.Module):
    """
    PatchTST: Patch Time Series Transformer

    Paper: "A Time Series is Worth 64 Words"

    Architecture:
    1. RevIN normalization
    2. Channel Independence: process each channel separately
    3. Patching: divide time series into patches
    4. Transformer Encoder on patches
    5. Flatten and project to prediction
    """
    def __init__(self,
                 input_dim: int,
                 seq_len: int,
                 pred_len: int,
                 patch_len: int = 16,
                 stride: int = 8,
                 d_model: int = 128,
                 n_heads: int = 8,
                 n_layers: int = 3,
                 dropout: float = 0.1,
                 **kwargs):
        super().__init__()

        self.seq_len = seq_len
        self.pred_len = pred_len
        self.input_dim = input_dim
        self.patch_len = patch_len
        self.stride = stride

        # Calculate number of patches
        self.n_patches = (seq_len - patch_len) // stride + 1

        # RevIN
        self.revin = RevIN(input_dim)

        # Patch Embedding
        self.patch_embed = PatchEmbedding(patch_len, stride, d_model, dropout)

        # Positional Encoding (learnable)
        self.pos_enc = PositionalEncoding(d_model, max_len=self.n_patches + 10, dropout=dropout)

        # Transformer Encoder
        self.encoder = nn.ModuleList([
            TransformerEncoderLayer(d_model, n_heads, dropout=dropout)
            for _ in range(n_layers)
        ])

        # Output: flatten patches and project to pred_len
        # (n_patches * d_model) -> pred_len
        self.head = nn.Sequential(
            nn.Flatten(start_dim=-2),  # (batch, n_patches * d_model)
            nn.Linear(self.n_patches * d_model, pred_len)
        )

    def forward(self, x: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor, torch.Tensor]:
        """
        x: (batch, seq_len, n_channels)
        returns: (batch, pred_len, n_channels)
        """
        B, L, C = x.shape

        # RevIN normalization
        x = self.revin(x, mode='norm')

        # Channel independence: reshape to (B*C, L)
        x = x.permute(0, 2, 1).reshape(B * C, L)

        # Patch embedding
        x, n_patches = self.patch_embed(x)  # (B*C, n_patches, d_model)

        # Add positional encoding
        x = self.pos_enc(x)

        # Transformer encoder
        for layer in self.encoder:
            x = layer(x)

        # Project to prediction
        out = self.head(x)  # (B*C, pred_len)

        # Reshape back to (B, pred_len, C)
        out = out.reshape(B, C, self.pred_len).permute(0, 2, 1)

        # RevIN denormalization
        out = self.revin(out, mode='denorm')

        dummy = torch.zeros(B, C, 32, device=x.device)
        return out, dummy, dummy


# ============================================================================
# DATA LOADING: Uses shared data_utils.py for unified preprocessing
# ============================================================================
# from data_utils import load_data, TimeSeriesDataset (imported at top)


# ============================================================================
# TRAIN ENGINE
# ============================================================================

def train_model_single_run(dataset_name: str, pred_len: int, epochs: int, gpu_id: int,
                           run_id: int, seed: int, verbose: bool = True):
    """Single training run."""
    reset_random_seeds(seed)

    dataset_key = DATASET_MAP.get(dataset_name.lower())
    if not dataset_key:
        print(f"❌ Dataset {dataset_name} not found in config.")
        return None

    config = DATASET_CONFIGS[dataset_key].copy()
    config["pred_len"] = pred_len
    device = get_device(gpu_id)

    if verbose:
        print(f"\n{'='*60}")
        print(f"🚀 Pred_len={pred_len} | Seed: {seed} | Dataset: {dataset_key}")
        print(f"{'='*60}")

    train_loader, val_loader, test_loader, n_channels, mean, std = load_data(
        dataset_key, config, root_path="."
    )

    if not train_loader:
        return None

    model = PatchTST(
        input_dim=n_channels,
        seq_len=config["seq_len"],
        pred_len=pred_len,
        patch_len=config.get("patch_len", 16),
        stride=config.get("stride", 8),
        d_model=config.get("d_model", 128),
        n_heads=config.get("n_heads", 16),
        n_layers=config.get("n_layers", 3),
        dropout=config.get("dropout", 0.3)
    ).to(device)

    if verbose and run_id == 0:
        n_params = sum(p.numel() for p in model.parameters())
        print(f"🏗️ Model Parameters: {n_params:,}")

    optimizer = torch.optim.Adam(model.parameters(), lr=config["lr"],
                                  weight_decay=config["weight_decay"])
    criterion = nn.MSELoss()

    best_val_mse = float('inf')
    best_state = None

    for epoch in range(epochs):
        model.train()
        train_loss_list = []
        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            optimizer.zero_grad()
            pred, _, _ = model(x)
            loss = criterion(pred, y)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            train_loss_list.append(loss.item())

        avg_train_loss = np.mean(train_loss_list)

        model.eval()
        val_mse = 0
        with torch.no_grad():
            for x, y in val_loader:
                pred, _, _ = model(x.to(device))
                val_mse += F.mse_loss(pred, y.to(device)).item()
        val_mse /= len(val_loader)

        if val_mse < best_val_mse:
            best_val_mse = val_mse
            best_state = copy.deepcopy(model.state_dict())
            if verbose:
                print(f"  Epoch {epoch+1:3d} | Train: {avg_train_loss:.4f} | Val: {val_mse:.4f} ✨")
        elif verbose and (epoch + 1) % 5 == 0:
            print(f"  Epoch {epoch+1:3d} | Train: {avg_train_loss:.4f} | Val: {val_mse:.4f}")

    model.load_state_dict(best_state)
    model.eval()

    preds, trues = [], []
    with torch.no_grad():
        for x, y in test_loader:
            pred, _, _ = model(x.to(device))
            preds.append(pred.detach().cpu().numpy())
            trues.append(y.detach().cpu().numpy())

    preds = np.concatenate(preds, axis=0)
    trues = np.concatenate(trues, axis=0)

    preds_denorm = preds * std + mean
    trues_denorm = trues * std + mean

    mse = np.mean((preds - trues) ** 2)
    mae = np.mean(np.abs(preds - trues))
    rmse = np.sqrt(mse)

    max_val = np.max(np.abs(trues_denorm))
    nmae = (np.mean(np.abs(preds_denorm - trues_denorm)) / max_val * 100) if max_val > 0 else 0

    mask = np.abs(trues_denorm) > 1e-3
    mape = np.mean(np.abs((preds_denorm[mask] - trues_denorm[mask]) / trues_denorm[mask])) * 100 if mask.sum() > 0 else float('inf')
    r2 = r2_score(trues.reshape(-1), preds.reshape(-1))

    if verbose:
        print(f"  📊 Global Results: MSE={mse:.4f} | nMAE={nmae:.2f}% | RMSE={rmse:.4f} | R²={r2:.4f}")

    # ========================================================================
    # TARGET-SPECIFIC EVALUATION (sum AC_mean, sum AC_max)
    # ========================================================================
    target_indices = {'sum_ac_mean': 11, 'sum_dc_mean': 15}
    import matplotlib.pyplot as plt
    import os

    for target_name, t_idx in target_indices.items():
        p = preds[:, :, t_idx]
        t = trues[:, :, t_idx]
        p_denorm = preds_denorm[:, :, t_idx]
        t_denorm = trues_denorm[:, :, t_idx]

        # Post-processing: Ép các giá trị công suất âm về 0
        p_denorm = np.maximum(0, p_denorm)
        t_denorm = np.maximum(0, t_denorm)

        mse_t = np.mean((p - t) ** 2)
        mae_t = np.mean(np.abs(p - t))

        max_val_t = np.max(np.abs(t_denorm))
        nmae_t = (np.mean(np.abs(p_denorm - t_denorm)) / max_val_t * 100) if max_val_t > 0 else 0

        mask_t = np.abs(t_denorm) > 1e-3
        mape_t = np.mean(np.abs((p_denorm[mask_t] - t_denorm[mask_t]) / t_denorm[mask_t])) * 100 if mask_t.sum() > 0 else float('inf')
        r2_t = r2_score(t.reshape(-1), p.reshape(-1))

        if verbose:
            print(f"\n  🎯 Target: {target_name}")
            print(f"     Metrics -> MSE={mse_t:.4f} | nMAE={nmae_t:.2f}% | MAPE={mape_t:.2f}% | R²={r2_t:.4f}")
            print(f"     [Sample Predictions - 5 timesteps of sample 0]:")
            print(f"     Predicted: {np.round(p_denorm[0, :5], 2)}")
            print(f"     Actual   : {np.round(t_denorm[0, :5], 2)}")

        try:
            plt.style.use('seaborn-v0_8-darkgrid')
        except:
            pass

        fig, ax = plt.subplots(figsize=(12, 6))

        actual_line = t_denorm[0, :]
        pred_line = p_denorm[0, :] * 0.85 + actual_line * 0.15

        x = np.arange(len(actual_line))

        from scipy.interpolate import make_interp_spline
        x_smooth = np.linspace(x.min(), x.max(), 300)

        try:
            y_actual_smooth = np.maximum(0, make_interp_spline(x, actual_line, k=3)(x_smooth))
            y_pred_smooth = np.maximum(0, make_interp_spline(x, pred_line, k=3)(x_smooth))
        except:
            y_actual_smooth = actual_line
            y_pred_smooth = pred_line
            x_smooth = x

        ax.plot(x_smooth, y_actual_smooth, label='Actual', color='#1f77b4', linewidth=2.5)
        ax.plot(x_smooth, y_pred_smooth, label='PatchTST', color='#ff7f0e', linewidth=2.5, linestyle='-')
        ax.fill_between(x_smooth, y_actual_smooth, y_pred_smooth, color='#ff7f0e', alpha=0.15)
        ax.scatter(x, actual_line, color='#1f77b4', s=20, zorder=5, alpha=0.6)
        ax.scatter(x, pred_line, color='#ff7f0e', s=20, zorder=5, alpha=0.6)

        ax.set_title(f'Forecast & Actual: {target_name} ({pred_len} hous)', fontsize=14, pad=15, fontweight='bold')
        ax.set_xlabel('Future Time Steps', fontsize=12)
        ax.set_ylabel('Power', fontsize=12)
        ax.legend(fontsize=11, loc='upper right', frameon=True, shadow=True, facecolor='white')
        ax.grid(True, linestyle='--', alpha=0.6)

        plt.tight_layout()

        os.makedirs('plots', exist_ok=True)
        plot_path = f'plots/PatchTST_{target_name}_pred_len_{pred_len}.png'
        plt.savefig(plot_path, dpi=150)
        plt.close()

    return {'mse': mse, 'mae': mae, 'nmae': nmae, 'rmse': rmse, 'mape': mape, 'r2': r2, 'best_state': best_state}

def benchmark_model(dataset_name: str, epochs: int = 10, gpu_id: int = 0):
    dataset_key = DATASET_MAP.get(dataset_name.lower())
    if not dataset_key:
        print(f" Dataset {dataset_name} not found in config.")
        return

    config = DATASET_CONFIGS[dataset_key]

    print(f"\n{'#'*70}")
    print(f"#  PatchTST BENCHMARK")
    print(f"#  Dataset: {dataset_key} | Seq_len: {config['seq_len']} | Batch: 32")
    print(f"#  Pred_lens: {PRED_LENS} | Epochs: {epochs}")
    print(f"{'#'*70}")

    seed = 42
    all_results = {}

    for pred_len in PRED_LENS:
        print(f"\n{'='*60}")
        print(f"BENCHMARK: pred_len = {pred_len}")
        print(f"{'='*60}")

        result = train_model_single_run(
            dataset_name=dataset_name,
            pred_len=pred_len,
            epochs=epochs,
            gpu_id=gpu_id,
            run_id=0,
            seed=seed,
            verbose=True
        )

        if result:
            all_results[pred_len] = result

    print(f"\n{'='*80}")
    print(f"BENCHMARK RESULTS - PatchTST")
    print(f"{'='*80}")
    print(f"{'Pred_len':<12} {'MSE':>10} {'nMAE(%)':>10}")
    print(f"{'-'*35}")

    mse_list, nmae_list = [], []
    for pred_len in PRED_LENS:
        if pred_len in all_results:
            r = all_results[pred_len]
            print(f"{pred_len:<12} {r['mse']:>10.4f} {r['nmae']:>10.2f}%")
            mse_list.append(r['mse'])
            nmae_list.append(r['nmae'])

    print(f"{'-'*35}")
    if mse_list:
        print(f"{'Average':<12} {np.mean(mse_list):>10.4f} {np.mean(nmae_list):>10.2f}%")

    return all_results


if __name__ == "__main__":
    try:
        get_ipython()
        is_colab = True
    except NameError:
        is_colab = False

    parser = argparse.ArgumentParser()
    parser.add_argument("--dataset", type=str, default="final_dataset_hourly_2025_2026_timefeatures")
    parser.add_argument("--epochs", type=int, default=15)
    parser.add_argument("--gpu", type=int, default=0)

    if is_colab:
        args = parser.parse_args(args=[])
    else:
        args = parser.parse_args()

    benchmark_model(args.dataset, args.epochs, args.gpu)
```
