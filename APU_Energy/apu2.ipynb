```python
energy_new = energy.drop_duplicates().reset_index(drop=True)
```

```python
energy_new.describe()
```

```python
energy_apu2 = energy_new.loc[(energy_new['System'] == 'APU 2')]
```

```python
energy_apu2.info()
```

```python
metadata_df = energy_apu2.iloc[:, :3]
numeric_df = energy_apu2.loc[:, 'CH 1 pos/Ah':].astype(float).diff()
energy_apu2_perminute = pd.concat([metadata_df, numeric_df], axis=1).dropna()
```

```python
energy_apu2_perminute
```

```python
energy_apu2_perminute.describe()
```

```python
energy_apu2_perminute.info()
```

```python
apu2 = energy_apu2_perminute.copy()
# 1. Đảm bảo cột Time Stamp là kiểu datetime và đặt làm Index
apu2['TimeStamp'] = pd.to_datetime(apu2['TimeStamp'])
apu2.set_index('TimeStamp', inplace=True)

# 2. Sau đó mới chạy lại lệnh vẽ biểu đồ
cols_to_plot = ['CH 1 pos/Ah']
apu2[cols_to_plot].resample('D').mean().plot(subplots=True, figsize=(16, 6), grid=True)
#plt.ylim(350,400)
plt.show()
```

```python
cols_to_plot = ['CH 2 pos/Ah']
apu2[cols_to_plot].resample('D').mean().plot(subplots=True, figsize=(16, 6), grid=True)
#plt.ylim(350,400)
plt.show()
```
