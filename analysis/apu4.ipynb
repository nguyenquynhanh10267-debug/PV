Có ngày APU 4 bị mất data

```python
apu4 = stats_apu4.copy()
# 1. Đảm bảo cột Time Stamp là kiểu datetime và đặt làm Index
apu4['Time Stamp'] = pd.to_datetime(apu4['Time Stamp'])
apu4.set_index('Time Stamp', inplace=True)

# 2. Sau đó mới chạy lại lệnh vẽ biểu đồ
cols_to_plot = ['VL1N/V']
apu4[cols_to_plot].resample('D').mean().plot(subplots=True, figsize=(16, 6), grid=True)
#plt.ylim(350,400)
plt.show()
```
